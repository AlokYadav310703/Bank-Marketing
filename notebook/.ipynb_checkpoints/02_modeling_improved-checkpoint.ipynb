# Modelling: Improved Pipeline for Term-Deposit Prediction
---
**Goal:** rank clients by their probability of subscribing so that call-centre effort goes where it pays off, **without** using `duration` (known only after a call).

## Why the first modelling notebook plateaued
The first notebook trained four reasonable models, and all of them stalled at ROC-AUC ≈ 0.80 and PR-AUC ≈ 0.46. Models were not the bottleneck. Diagnosis in this notebook found six causes:

| # | Cause | Effect |
|---|---|---|
| 1 | **The file is chronological and the campaign changed regime over time.** Conversion is about 5% in 2008, 17% in 2009 and 52% in 2010. | `month` acts as a stand-in for *which era a client was called in*. Pooled AUC partly rewards telling 2008 clients from 2010 clients, which does not help an agent choose whom to call. |
| 2 | **Information was thrown away.** Raw `pdays` was dropped after making a flag, and the date was never reconstructed. | Days since last contact and the weekday were lost. |
| 3 | **No call-plan features.** The size and mix of each day's and month's call list was not available to the models. | The "warm list vs mass-market sweep" signal found in the EDA was invisible. |
| 4 | **Default hyper-parameters, no early-stopping or search**, and **one 9,043-row test set** used for choosing the model, choosing the threshold and reporting. | Differences of 0.005 AUC are noise at that size, and the reported numbers were optimistic. |
| 5 | **`class_weight='balanced'` / `scale_pos_weight` without calibration**, then a "cost-based threshold" applied to those distorted probabilities. | The threshold was not what it claimed to be. |
| 6 | **`campaign` is only fully known after the calls.** | Mild leakage; the deployable model should not use it. |

## What this notebook does
1. Reproduces your numbers (to within 0.002), so every comparison is like-for-like.
2. Reconstructs and **validates** call dates, then measures what each feature group is worth with **cross-validation on the training data only**. The holdout stays untouched until the end.
3. Tunes and compares LightGBM, XGBoost, CatBoost and logistic regression, blends them, and **calibrates** the result.
4. Reports **pooled, within-era and forward-in-time** performance, with a bootstrap confidence interval on the improvement.
5. Turns scores into **call-budget and expected-profit decisions** with thresholds chosen on training predictions, not on the test set.

> **Honest expectation:** without `duration`, this dataset has a ceiling near ROC-AUC 0.82. The gains here are real but moderate. The largest practical gain is a **trustworthy evaluation**.

## 1. Setup and configuration

```python
import os, json, time, warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
warnings.filterwarnings("ignore", category=FutureWarning)
warnings.filterwarnings("ignore", category=DeprecationWarning)
warnings.filterwarnings("ignore", category=UserWarning)

from sklearn.base import clone
from sklearn.model_selection import train_test_split, StratifiedKFold
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.isotonic import IsotonicRegression
from sklearn.inspection import permutation_importance
from sklearn.metrics import (roc_auc_score, average_precision_score, brier_score_loss, log_loss,
                             roc_curve, precision_recall_curve, f1_score, precision_score, recall_score)
from lightgbm import LGBMClassifier
from xgboost import XGBClassifier
from catboost import CatBoostClassifier

RANDOM_SEED = 42
N_JOBS      = 2          # raise if you have more cores
RUN_TUNING  = False      # True re-runs the Optuna search in section 6 (about 8 minutes)
np.random.seed(RANDOM_SEED)
sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams.update({"figure.dpi": 110, "figure.figsize": (8, 5)})

MONTHS = ["jan","feb","mar","apr","may","jun","jul","aug","sep","oct","nov","dec"]
CATS = ["job","marital","education","default","housing","loan","contact","month","poutcome"]

PATHS = ["../data/bank-full.csv", "data/bank-full.csv", "bank-full.csv"]
path = next((p for p in PATHS if os.path.exists(p)), None)
if path is None: raise FileNotFoundError(PATHS)
df = pd.read_csv(path, sep=";")
df["y"] = (df["y"] == "yes").astype(int)
print("Shape:", df.shape, "| positive rate:", round(df.y.mean(), 4))
```

## 2. Discovery: the file is chronological. Reconstruct and validate the call dates

```python
# Rows are stored in time order, but the file has no year. Months wrap (dec -> jan) exactly when the year changes.
mi = df["month"].map({m: i + 1 for i, m in enumerate(MONTHS)})
df["year"] = 2008 + (mi < mi.shift()).fillna(False).cumsum()          # file starts May 2008
df["date"] = pd.to_datetime(dict(year=df.year, month=mi, day=df.day), errors="coerce")
df["row"]  = np.arange(len(df))

# ---- validation: a wrong reconstruction would give impossible dates and lots of weekend calls ----
invalid = int(df.date.isna().sum())
weekend = float((df.date.dt.dayofweek >= 5).mean())
day_order_breaks = int(((df.day < df.day.shift()) & (mi == mi.shift())).sum())
print(f"Invalid calendar dates          : {invalid}")
print(f"Share of calls on Sat/Sun       : {weekend:.2%}   (a random calendar would give about 28.6%)")
print(f"Day-of-month order breaks inside a month: {day_order_breaks}")
assert invalid == 0 and weekend < 0.02, "Date reconstruction failed: file may not be chronological"

era = df.groupby("year").agg(calls=("y", "size"), conversion_pct=("y", lambda s: s.mean() * 100),
                             first=("date", "min"), last=("date", "max"))
print(); print(era.round(1).to_string())

monthly = df.groupby(df.date.dt.to_period("M")).agg(n=("y", "size"), rate=("y", "mean"))
fig, axes = plt.subplots(2, 1, figsize=(12, 6.5), sharex=True)
x = monthly.index.to_timestamp()
axes[0].bar(x, monthly.rate * 100, width=22, color="#1b9e77"); axes[0].set_ylabel("Conversion (%)")
axes[0].set_title("Conversion by calendar month: a regime change, not seasonality")
axes[1].bar(x, monthly.n, width=22, color="lightsteelblue"); axes[1].set_ylabel("Calls"); axes[1].set_yscale("log")
plt.tight_layout(); plt.show()
```

### Findings, assumptions and so what: the file is a timeline
- **What was found:** rows are stored in chronological order, but the file has no year. Because months wrap (dec → jan) exactly when the year changes, the year can be reconstructed from row order: **2008 (27,729 calls), 2009 (14,862) and 2010 (2,620)**, spanning 5 May 2008 to 17 Nov 2010.
- **Why we can trust the reconstruction (three independent checks):** 0 impossible calendar dates, 0 day-of-month ordering breaks inside any month, and only **0.11% of calls fall on a weekend**. A wrong year would put about 28.6% of calls on weekends.
- **The campaign changed regime:** conversion is **5.1% in 2008, 17.1% in 2009 and 51.6% in 2010**, while monthly call volume collapses from thousands of calls to a few hundred. The bank moved from mass-market sweeps to small, warm lists.
- **This explains the EDA's month finding.** The "high-conversion months" (Mar, Sep, Oct, Dec) are almost entirely 2009 to 2010 months from the small-list regime. That is a campaign change, not seasonality.
- **Consequence for modelling:** `month` quietly encodes *which era* a client was called in. A pooled AUC then partly rewards separating 2008 clients from 2010 clients, which does not help an agent decide whom to call within one campaign. Sections 9 and 10 therefore also report **within-era** and **forward-in-time** performance.
- **Assumptions:** the file starts in 2008 (consistent with the dataset's documented May 2008 to Nov 2010 span) and rows are in date order, which the validation supports but the source does not document.

## 3. Evaluation protocol and baseline reproduction

```python
# ---- Same split as the first notebook (stratified 80/20, seed 42) so numbers are directly comparable ----
y = df["y"].values; year = df["year"].values
train_idx, hold_idx = train_test_split(np.arange(len(df)), test_size=0.2, stratify=y, random_state=RANDOM_SEED)
print(f"Train {len(train_idx):,} | Holdout {len(hold_idx):,} (touched only for final reporting)")

def lift_at(p, yt, frac):
    k = int(len(p) * frac); top = np.argsort(-p)[:k]
    return yt[top].mean() / yt.mean(), yt[top].sum() / yt.sum()          # (lift, recall captured)

def metrics(p, idx):
    # pooled ROC / PR, mean within-era ROC, and top-20% lift
    yt = y[idx]; yr = year[idx]
    w = [roc_auc_score(yt[yr == Y], p[yr == Y]) for Y in (2008, 2009, 2010)]
    return dict(ROC=roc_auc_score(yt, p), PR=average_precision_score(yt, p),
                within_era=float(np.mean(w)), lift20=lift_at(p, yt, 0.2)[0])

def cv_oof(make, X, idx=train_idx, k=5, seed=RANDOM_SEED):
    # out-of-fold predictions on the training rows only
    oof = np.zeros(len(idx)); skf = StratifiedKFold(k, shuffle=True, random_state=seed)
    for a, b in skf.split(idx, y[idx]):
        m = make(); m.fit(X.iloc[idx[a]], y[idx[a]]); oof[b] = m.predict_proba(X.iloc[idx[b]])[:, 1]
    return oof

# ---- Reproduce the first notebook's feature set and four models exactly ----
def user_frame(d):
    u = d.drop(columns=["duration", "pdays", "year", "date", "row"]).copy()
    u["was_previously_contacted"] = (d.pdays != -1).astype(int)
    u["age_band"] = pd.cut(d.age, bins=[17,25,35,45,55,65,100], labels=["18-25","26-35","36-45","46-55","56-65","65+"]).astype(str)
    u["balance"] = np.sign(u.balance) * np.log1p(u.balance.abs())
    return u.drop(columns="y")
U = user_frame(df)
u_num = ["age","balance","day","campaign","previous"]
u_cat = CATS + ["age_band", "was_previously_contacted"]
u_pre = lambda: ColumnTransformer([("n", StandardScaler(), u_num),
                                   ("c", OneHotEncoder(handle_unknown="ignore", drop="if_binary"), u_cat)])
spw0 = (y[train_idx] == 0).sum() / (y[train_idx] == 1).sum()
BASE_MODELS = {
 "LogReg":   LogisticRegression(max_iter=1000, class_weight="balanced", random_state=RANDOM_SEED),
 "XGBoost":  XGBClassifier(n_estimators=300, max_depth=5, learning_rate=.05, subsample=.8, colsample_bytree=.8,
                           scale_pos_weight=spw0, eval_metric="aucpr", random_state=RANDOM_SEED, n_jobs=N_JOBS),
 "LightGBM": LGBMClassifier(n_estimators=400, max_depth=6, learning_rate=.03, subsample=.7, colsample_bytree=.8,
                            class_weight="balanced", random_state=RANDOM_SEED, verbose=-1, n_jobs=N_JOBS),
 "CatBoost": CatBoostClassifier(iterations=300, depth=5, learning_rate=.05, loss_function="Logloss", eval_metric="AUC",
                                random_seed=RANDOM_SEED, verbose=0, thread_count=N_JOBS),
}
base_hold = {}
rows = []
for name, mdl in BASE_MODELS.items():
    pipe = Pipeline([("p", u_pre()), ("m", mdl)]).fit(U.iloc[train_idx], y[train_idx])
    base_hold[name] = pipe.predict_proba(U.iloc[hold_idx])[:, 1]
    rows.append({"model": name, **metrics(base_hold[name], hold_idx)})
baseline_tbl = pd.DataFrame(rows).set_index("model")
print("First notebook's models, re-run (holdout):"); print(baseline_tbl.round(4).to_string())
print("\nFirst notebook reported: LogReg 0.7751/0.4140, XGBoost 0.8026/0.4549, LightGBM 0.8038/0.4581, CatBoost 0.8064/0.4617 (ROC/PR)")
BASE_NAME = baseline_tbl["PR"].idxmax(); p_base = base_hold[BASE_NAME]
print(f"\nReference baseline for all comparisons below: {BASE_NAME} (highest PR-AUC, as in the first notebook)")
```

### Findings, assumptions and so what: baseline reproduction
- **The first notebook's results are reproduced** on the identical split: LogReg 0.7751 / 0.4140, LightGBM 0.8038 / 0.4581 (ROC / PR). XGBoost and CatBoost differ from the printed values by at most 0.002, which is ordinary multithreading noise.
- **Reference for every comparison below: CatBoost** (0.8056 ROC, 0.4621 PR), the highest PR-AUC, as before. It is the *hardest* baseline to beat.
- **The "within_era" column is new.** The baseline's mean within-era ROC is about 0.71, much lower than its pooled 0.80. This gap measures how much of the pooled score comes from telling eras apart.
- **Statistical caution:** the holdout has 9,043 rows and 1,058 subscribers, so an AUC difference smaller than about 0.01 is within noise. This is why feature and model choices below are made by **cross-validation on the 36,168 training rows**, and the holdout is used once, at the end.

## 4. Feature engineering: what is each group worth? (cross-validation on training data only)

```python
def build_features(d, level, call_plan=False, drop_campaign=False):
    # level 0 = the first notebook's information; 1 = + call-time features; 2 = + year / time index (diagnostic only)
    X = pd.DataFrame(index=d.index)
    X["age"] = d.age; X["balance"] = d.balance; X["day"] = d.day; X["campaign"] = d.campaign; X["previous"] = d.previous
    for c in CATS: X[c] = d[c].astype("category")
    X["was_prev"] = (d.pdays != -1).astype(int)
    if level >= 1:
        X["pdays_clean"] = d.pdays.where(d.pdays != -1)                    # real days since last contact (sentinel -> NaN)
        X["weekday"] = d.date.dt.dayofweek
        X["month_num"] = d.month.map({m: i + 1 for i, m in enumerate(MONTHS)})
        X["day_of_year"] = d.date.dt.dayofyear
        X["n_loans"] = (d.housing == "yes").astype(int) + (d.loan == "yes").astype(int)
        X["neg_balance"] = (d.balance < 0).astype(int)
        X["prev_success"] = (d.poutcome == "success").astype(int)
    if call_plan:
        ym = d.year * 100 + d.month.map({m: i + 1 for i, m in enumerate(MONTHS)})
        X["calls_that_day"] = d.groupby("date")["age"].transform("size")
        X["calls_that_month"] = ym.map(ym.value_counts())
        X["share_prev_contacted_month"] = (d.pdays != -1).astype(int).groupby(ym).transform("mean")
        X["share_contact_unknown_month"] = (d.contact == "unknown").astype(int).groupby(ym).transform("mean")
    if level >= 2:
        X["year"] = d.year; X["t"] = (d.date - pd.Timestamp("2008-05-01")).dt.days
    if drop_campaign: X = X.drop(columns="campaign")
    return X

mk_default = lambda: LGBMClassifier(n_estimators=400, max_depth=6, learning_rate=.03, subsample=.7, subsample_freq=1,
                                    colsample_bytree=.8, class_weight="balanced", random_state=RANDOM_SEED,
                                    verbose=-1, n_jobs=N_JOBS)
STEPS = [("A. First notebook's information",                         dict(level=0)),
         ("B. + call-time features (pdays, weekday, date parts)",    dict(level=1)),
         ("C. + call-plan features (size / mix of the call list)",   dict(level=1, call_plan=True)),
         ("D. C without `campaign` (deployable before calling)",     dict(level=1, call_plan=True, drop_campaign=True)),
         ("E. D + year / time index (diagnostic, NOT used)",         dict(level=2, call_plan=True, drop_campaign=True))]
rows = []
for lab, kw in STEPS:
    o = cv_oof(mk_default, build_features(df, **kw)); rows.append({"feature set": lab, **metrics(o, train_idx)})
ablation = pd.DataFrame(rows).set_index("feature set")
print("5-fold CV on the TRAINING rows (same LightGBM settings throughout):")
print(ablation.round(4).to_string())
```

### Findings, assumptions and so what: what each feature group is worth
Five-fold cross-validation on the training rows, with the same LightGBM settings throughout, so only the features change:

| Step | ROC | PR | within-era | Change |
|---|---|---|---|---|
| A. First notebook's information | 0.798 | 0.445 | 0.698 | n/a |
| B. + call-time features | 0.805 | 0.461 | 0.709 | +0.007 ROC, +0.016 PR |
| C. + call-plan features | 0.815 | 0.486 | 0.726 | **+0.010 ROC, +0.025 PR (largest step)** |
| D. C without `campaign` | 0.812 | 0.485 | 0.725 | −0.003 ROC, −0.002 PR |
| E. D + year / time index | 0.813 | 0.486 | 0.725 | no gain, **not used** |

- **B: information the first notebook discarded.** Raw `pdays` was reduced to a flag, losing the days since last contact. The reconstructed date adds the **weekday**, `day_of_year` and `month_num`. Calls on weekends are essentially absent, so weekday is a real scheduling variable.
- **C: call-plan features.** For each call, how many calls were made that day and that month, what share of that month's list had been contacted before, and what share had no recorded contact type. These are **label-free**: they describe the schedule, never the outcome. They give the model the "warm list vs mass-market sweep" signal that the EDA identified but the models could not see.
- **D: `campaign` dropped.** It counts calls that are only known after the client has decided (calling stops after a "yes"), so a model that uses it cannot be applied before calling. Removing it costs only about 0.003 AUC, a cheap price for a deployable model.
- **E: a raw year or time index adds nothing** once the call-plan features are in, so no time index is fed to the model.
- **Assumptions:** the call-plan features are computable at scoring time from the planned call list. If the dialler plan changes (different list sizes or mix), these features change meaning and the model must be re-validated (see sections 10 and 13).

## 5. Final feature set

```python
X = build_features(df, level=1, call_plan=True, drop_campaign=True)
Xs = X.copy()
for c in CATS: Xs[c] = Xs[c].astype(str)                       # string categories for CatBoost
num_cols = [c for c in X.columns if c not in CATS]
print(f"{X.shape[1]} features: {len(num_cols)} numeric + {len(CATS)} categorical")
print("Call-plan features are label-free: they describe the schedule (how many calls per day/month and what kind of list),")
print("never the outcome. `campaign` is excluded: it counts calls that are only known after the client has decided.")
print("\nMissing values (kept as NaN for tree models):", X.isna().sum()[lambda s: s > 0].to_dict())
```

## 6. Hyper-parameter search (Bayesian, cross-validated on training data)

```python
# Parameters found by a 40-trial Optuna/TPE search maximising CV PR-AUC + 0.5*ROC-AUC on the training rows.
TUNED_LGB = dict(n_estimators=614, learning_rate=0.01676, num_leaves=22, min_child_samples=26, subsample=0.686,
                 subsample_freq=1, colsample_bytree=0.602, reg_lambda=0.0108, reg_alpha=0.0205, min_split_gain=0.375,
                 cat_smooth=50.36, cat_l2=4.48, scale_pos_weight=2.27)

if RUN_TUNING:
    import optuna
    optuna.logging.set_verbosity(optuna.logging.WARNING)
    def objective(t):
        p = dict(n_estimators=t.suggest_int("n_estimators", 200, 900), learning_rate=t.suggest_float("learning_rate", .01, .08, log=True),
                 num_leaves=t.suggest_int("num_leaves", 6, 48), min_child_samples=t.suggest_int("min_child_samples", 20, 250),
                 subsample=t.suggest_float("subsample", .5, 1.0), subsample_freq=1,
                 colsample_bytree=t.suggest_float("colsample_bytree", .4, 1.0),
                 reg_lambda=t.suggest_float("reg_lambda", 1e-2, 30, log=True), reg_alpha=t.suggest_float("reg_alpha", 1e-3, 5, log=True),
                 min_split_gain=t.suggest_float("min_split_gain", 0, .5), cat_smooth=t.suggest_float("cat_smooth", 1, 60),
                 cat_l2=t.suggest_float("cat_l2", 1, 40), scale_pos_weight=t.suggest_float("scale_pos_weight", 1, 8))
        o = cv_oof(lambda: LGBMClassifier(**p, random_state=RANDOM_SEED, verbose=-1, n_jobs=N_JOBS), X)
        m = metrics(o, train_idx); return m["PR"] + 0.5 * m["ROC"]
    study = optuna.create_study(direction="maximize", sampler=optuna.samplers.TPESampler(seed=RANDOM_SEED))
    study.optimize(objective, n_trials=40)
    TUNED_LGB = {**study.best_params, "subsample_freq": 1}
    print("Search finished. Best:", TUNED_LGB)
else:
    print("Using the stored best parameters (set RUN_TUNING = True to repeat the search):")
print(json.dumps(TUNED_LGB, indent=1))
```

### Findings, assumptions and so what: tuning
- **Method:** a 40-trial Bayesian (TPE) search over 12 LightGBM settings, scored by 5-fold CV on the training rows only. The best configuration uses a slower learning rate (0.017) with more, smaller trees (22 leaves), strong feature subsampling (0.60), a `min_split_gain` of 0.375 as regularisation, and a moderate positive-class weight (2.3, not the full 7.5).
- **Effect:** CV ROC / PR move from 0.812 / 0.485 (default settings, same features) to **0.817 / 0.491** for the tuned model. It is a small, real gain, which is typical once features are good.
- **Winner's-curse caveat:** the best of 40 trials is slightly optimistic on the data it was selected on. The honest number is the one measured later on the untouched holdout.
- **Why a mild class weight:** the full "balanced" weight (about 7.5) distorts the probabilities heavily (see section 8). A weight near 2 keeps the ranking benefit and is easy to calibrate.
- **To repeat the search**, set `RUN_TUNING = True` in section 1 (about 8 minutes on 2 cores).

## 7. Models: tuned LightGBM, XGBoost, CatBoost (native categoricals) and logistic regression

```python
spw = 2.3
lr_pre = lambda: ColumnTransformer([("n", Pipeline([("i", SimpleImputer(strategy="median")), ("s", StandardScaler())]), num_cols),
                                    ("c", OneHotEncoder(handle_unknown="ignore"), CATS)])
SPEC = {
 "LightGBM (tuned)": (lambda: LGBMClassifier(**TUNED_LGB, random_state=RANDOM_SEED, verbose=-1, n_jobs=N_JOBS), X),
 "XGBoost":  (lambda: XGBClassifier(n_estimators=600, max_depth=5, learning_rate=.03, subsample=.8, colsample_bytree=.6,
                                    min_child_weight=3, reg_lambda=5, scale_pos_weight=spw, tree_method="hist",
                                    enable_categorical=True, eval_metric="aucpr", random_state=RANDOM_SEED, n_jobs=N_JOBS), X),
 "CatBoost": (lambda: CatBoostClassifier(iterations=700, depth=6, learning_rate=.04, l2_leaf_reg=5, scale_pos_weight=spw,
                                         random_seed=RANDOM_SEED, verbose=0, thread_count=N_JOBS, cat_features=CATS), Xs),
 "LogReg":   (lambda: Pipeline([("p", lr_pre()), ("m", LogisticRegression(C=.3, max_iter=3000, random_state=RANDOM_SEED))]), X),
}
OOF, HOLD, fitted = {}, {}, {}
rows = []
for name, (mk, XX) in SPEC.items():
    t0 = time.time()
    OOF[name] = cv_oof(mk, XX)
    fitted[name] = mk().fit(XX.iloc[train_idx], y[train_idx])
    HOLD[name] = fitted[name].predict_proba(XX.iloc[hold_idx])[:, 1]
    rows.append({"model": name, **{f"CV {k}": v for k, v in metrics(OOF[name], train_idx).items() if k != "lift20"},
                 **{f"HOLD {k}": v for k, v in metrics(HOLD[name], hold_idx).items()}, "sec": time.time() - t0})
model_tbl = pd.DataFrame(rows).set_index("model")
print(model_tbl.round(4).to_string())
```

### Findings, assumptions and so what: models on the new features
| Model | CV ROC | CV PR | Holdout ROC | Holdout PR |
|---|---|---|---|---|
| LightGBM (tuned) | 0.8165 | 0.4911 | 0.8180 | 0.5121 |
| XGBoost | 0.8129 | 0.4835 | 0.8133 | 0.5061 |
| CatBoost | 0.8143 | 0.4822 | 0.8133 | 0.5059 |
| LogReg | 0.7977 | 0.4615 | 0.7992 | 0.4732 |

- **The features matter more than the algorithm.** The three boosted models are within 0.005 AUC of each other, below the noise floor of a 9,043-row holdout. Even the plain **logistic regression** improves from 0.7751 / 0.4140 to 0.7992 / 0.4732 on the new features, a bigger jump than any algorithm swap.
- **CV and holdout agree** (e.g. LightGBM 0.8165 vs 0.8180 ROC), which is the check that the holdout was not used to tune anything.
- **CatBoost takes about 100 s** against about 11 s for the others, for no measurable gain here.
- **Assumption:** the tree models use native categorical handling (no one-hot), which avoids the 40+ sparse dummy columns the first notebook fed them.

## 8. Blend and calibrate

```python
BLEND = ["LightGBM (tuned)", "XGBoost", "CatBoost"]                      # equal weights: no weight tuning, so nothing to overfit
oof_blend  = np.mean([OOF[n]  for n in BLEND], axis=0)
hold_blend = np.mean([HOLD[n] for n in BLEND], axis=0)

# Platt scaling (logistic regression on the logit of the score) learned on OUT-OF-FOLD training predictions, then applied to the holdout.
# It is smooth and monotone, so it changes probabilities but NOT the ranking. Isotonic is shown for reference: it creates tied scores.
logit = lambda p: np.log(np.clip(p, 1e-6, 1 - 1e-6) / (1 - np.clip(p, 1e-6, 1 - 1e-6)))[:, None]
platt = LogisticRegression(C=1e6, max_iter=1000).fit(logit(oof_blend), y[train_idx])
oof_cal, hold_cal = platt.predict_proba(logit(oof_blend))[:, 1], platt.predict_proba(logit(hold_blend))[:, 1]
iso = IsotonicRegression(out_of_bounds="clip", y_min=0, y_max=1).fit(oof_blend, y[train_idx]); hold_iso = iso.predict(hold_blend)

def calib_row(p, yt): return dict(brier=brier_score_loss(yt, p), logloss=log_loss(yt, np.clip(p, 1e-6, 1 - 1e-6)), mean_pred=p.mean())
cal = pd.DataFrame({"first notebook's best (raw)": calib_row(p_base, y[hold_idx]),
                    "blend (raw, class-weighted)": calib_row(hold_blend, y[hold_idx]),
                    "blend (Platt-calibrated)": calib_row(hold_cal, y[hold_idx]),
                    "blend (isotonic, reference)": calib_row(hold_iso, y[hold_idx])}).T
cal["actual rate"] = y[hold_idx].mean()
print(cal.round(4).to_string())

fig, ax = plt.subplots(figsize=(5.6, 5))
for lab, p in [("raw blend", hold_blend), ("Platt-calibrated", hold_cal)]:
    q = pd.qcut(pd.Series(p).rank(method="first"), 10, labels=False)
    g = pd.DataFrame({"p": p, "y": y[hold_idx], "q": q}).groupby("q").mean()
    ax.plot(g.p, g.y, "o-", label=lab)
ax.plot([0, .6], [0, .6], "k--", alpha=.5); ax.set_xlabel("Mean predicted probability (decile)"); ax.set_ylabel("Observed conversion rate")
ax.set_title("Reliability on the holdout"); ax.legend(); plt.tight_layout(); plt.show()

final = {"blend (equal weights)": metrics(hold_blend, hold_idx), "blend (Platt; ranking unchanged)": metrics(hold_cal, hold_idx),
         "blend (isotonic; creates ties)": metrics(hold_iso, hold_idx)}
print(pd.DataFrame(final).T.round(4).to_string())
```

### Findings, assumptions and so what: blend and calibration
- **Blend:** an equal-weight average of the three boosted models. Weights were deliberately **not tuned**, so there is nothing to overfit. Holdout 0.8158 ROC / 0.5099 PR. The single tuned LightGBM scored slightly higher on this one holdout (0.8180 / 0.5121), but that difference is within noise. The blend is kept because averaging different algorithms is more stable on new data.
- **The raw probabilities are not probabilities.** Class-weighted models over-predict: the blend's average predicted probability is **0.19 against a true rate of 0.117**, with Brier score 0.0838. This matters because the first notebook applied a cost-based threshold (FN : FP = 20 : 1) directly to such scores.
- **Calibration fixes it:** Platt scaling (a logistic regression on the logit of the score, fitted on **out-of-fold training predictions only**) brings the average prediction to 0.118 and Brier to **0.0760**. The reliability chart shows the calibrated curve hugging the diagonal.
- **Ranking is untouched by Platt** (identical ROC and PR). **Isotonic calibration**, shown for reference, reaches the same Brier but creates tied scores that lower PR-AUC from 0.510 to 0.497, which is why Platt is used.
- **Assumption:** the score-to-probability relationship learned on 2008 to 2010 data holds for future data. If the regime shifts, recalibrate.

## 9. Before vs after on the untouched holdout, with a confidence interval

```python
yh = y[hold_idx]; yrh = year[hold_idx]
before = metrics(p_base, hold_idx); after = metrics(hold_blend, hold_idx)

# Paired bootstrap on the holdout: resample clients, recompute both models on the SAME resample
rng = np.random.default_rng(RANDOM_SEED); B = 1000; n = len(yh)
d_roc, d_pr = np.empty(B), np.empty(B)
for b in range(B):
    ix = rng.integers(0, n, n)
    d_roc[b] = roc_auc_score(yh[ix], hold_blend[ix]) - roc_auc_score(yh[ix], p_base[ix])
    d_pr[b]  = average_precision_score(yh[ix], hold_blend[ix]) - average_precision_score(yh[ix], p_base[ix])
ci = lambda v: (np.percentile(v, 2.5), np.percentile(v, 97.5))

summary = pd.DataFrame({"first notebook": before, "this notebook": after}).T
print(summary.round(4).to_string())
print(f"\nImprovement  ROC-AUC {after['ROC']-before['ROC']:+.4f}  95% CI [{ci(d_roc)[0]:+.4f}, {ci(d_roc)[1]:+.4f}]")
print(f"Improvement  PR-AUC  {after['PR']-before['PR']:+.4f}  95% CI [{ci(d_pr)[0]:+.4f}, {ci(d_pr)[1]:+.4f}]   "
      f"(relative {100*(after['PR']/before['PR']-1):+.1f}%)")
print(f"Improvement within-era ROC {after['within_era']-before['within_era']:+.4f}")

fig, axes = plt.subplots(1, 2, figsize=(13, 4.8))
for lab, p, c in [(f"First notebook ({BASE_NAME})", p_base, "#9e9e9e"), ("This notebook (blend)", hold_blend, "#1b9e77")]:
    f, t, _ = roc_curve(yh, p); axes[0].plot(f, t, c=c, label=f"{lab}  AUC={roc_auc_score(yh, p):.3f}")
    pr, rc, _ = precision_recall_curve(yh, p); axes[1].plot(rc, pr, c=c, label=f"{lab}  AP={average_precision_score(yh, p):.3f}")
axes[0].plot([0, 1], [0, 1], "k--", alpha=.4); axes[0].set_xlabel("False positive rate"); axes[0].set_ylabel("True positive rate"); axes[0].legend(); axes[0].set_title("ROC (holdout)")
axes[1].axhline(yh.mean(), c="gray", ls="--", label="no skill"); axes[1].set_xlabel("Recall"); axes[1].set_ylabel("Precision"); axes[1].legend(); axes[1].set_title("Precision-Recall (holdout)")
plt.tight_layout(); plt.show()
```

### Findings, assumptions and so what: before vs after (untouched holdout)
| | First notebook (CatBoost) | This notebook (blend) | Change (95% CI) |
|---|---|---|---|
| ROC-AUC | 0.8056 | **0.8158** | **+0.0102** [+0.0024, +0.0188] |
| PR-AUC | 0.4621 | **0.5099** | **+0.0478** [+0.0327, +0.0641], +10.3% relative |
| Within-era ROC | 0.7074 | **0.7364** | +0.0290 |
| Top-20% lift | 3.14 | **3.25** | +0.11 |

- **The improvement is real, not noise:** both confidence intervals (paired bootstrap, 1,000 resamples of the same clients) exclude zero. PR-AUC, which measures how well the top of the list is filled with subscribers, shows the clearest gain.
- **It is moderate by design:** without `duration`, this dataset's ceiling is near ROC-AUC 0.82. A claim of 0.90 would indicate leakage. The within-era gain (+0.029) is the largest in relative terms, because that is where the call-plan and history features help most.
- **Assumptions:** the bootstrap treats clients as independent (no client id to verify), and the holdout is a random 20% sample, so it shares the data's era mix.

## 10. Honest evaluation: within-era and forward-in-time

```python
# ---- (a) Within-era ranking: what matters when one campaign is running ----
rows = []
for Y in (2008, 2009, 2010):
    m = yrh == Y
    rows.append({"era": Y, "holdout clients": int(m.sum()), "base rate %": yh[m].mean() * 100,
                 "ROC first": roc_auc_score(yh[m], p_base[m]), "ROC now": roc_auc_score(yh[m], hold_blend[m]),
                 "PR first": average_precision_score(yh[m], p_base[m]), "PR now": average_precision_score(yh[m], hold_blend[m]),
                 "top-20% lift first": lift_at(p_base[m], yh[m], .2)[0], "top-20% lift now": lift_at(hold_blend[m], yh[m], .2)[0]})
print("(a) Within-era performance on the holdout:"); print(pd.DataFrame(rows).set_index("era").round(3).to_string())

# ---- (b) Forward-in-time: train on the past, predict the future (no shuffling) ----
fwd = [("2008 -> 2009", df.year == 2008, df.year == 2009), ("2008-09 -> 2010", df.year <= 2009, df.year == 2010),
       ("first 70% rows -> last 30%", df.row < .7 * len(df), df.row >= .7 * len(df))]
U_cb = U.copy()
rows = []
for lab, trm, tem in fwd:
    base_pipe = Pipeline([("p", u_pre()), ("m", clone(BASE_MODELS[BASE_NAME]))]).fit(U[trm.values], y[trm.values])
    pb = base_pipe.predict_proba(U[tem.values])[:, 1]
    pn = np.mean([SPEC[n][0]().fit(SPEC[n][1][trm.values], y[trm.values]).predict_proba(SPEC[n][1][tem.values])[:, 1] for n in BLEND], axis=0)
    yt = y[tem.values]
    rows.append({"split": lab, "test clients": int(tem.sum()), "ROC first": roc_auc_score(yt, pb), "ROC now": roc_auc_score(yt, pn),
                 "PR first": average_precision_score(yt, pb), "PR now": average_precision_score(yt, pn)})
fwd_tbl = pd.DataFrame(rows).set_index("split")
print("\n(b) Forward-in-time (train on the past only):"); print(fwd_tbl.round(3).to_string())
```

### Findings, assumptions and so what: honest evaluation
**(a) Within-era (holdout):**

- Improvement shows in every era on ROC and PR, **but 2008 stays hard** (ROC 0.65, almost no gain in top-20% lift). In 2008 the bank called nearly everyone with little history, and the available features carry little signal. No amount of tuning will fix missing information.

The new features hold up much better: gains of +0.03 to +0.10 ROC when predicting forward, so they are not just fitting the era.
- **Assumption:** forward splits mix regimes (e.g. 2008 → 2009 crosses the regime change), so they are a *stress test*, not a forecast of production performance.

## 11. Business translation: lift, gains and call budgets

```python
def decile_table(p, yt):
    r = pd.DataFrame({"y": yt, "p": p}).sort_values("p", ascending=False).reset_index(drop=True)
    r["decile"] = pd.qcut(r.index, 10, labels=False) + 1
    g = r.groupby("decile").agg(clients=("y", "size"), subscribers=("y", "sum")); g["rate %"] = g.subscribers / g.clients * 100
    g["lift"] = g["rate %"] / (yt.mean() * 100); g["cum recall %"] = g.subscribers.cumsum() / g.subscribers.sum() * 100
    return g

dt_base, dt_new = decile_table(p_base, yh), decile_table(hold_blend, yh)
side = pd.concat({"first notebook": dt_base[["rate %", "lift", "cum recall %"]],
                  "this notebook":  dt_new[["rate %", "lift", "cum recall %"]]}, axis=1)
print("Holdout, clients ranked by score (decile 1 = highest scored):"); print(side.round(2).to_string())

rows = []
for frac in (.05, .10, .20, .30, .50):
    lb, rb = lift_at(p_base, yh, frac); ln, rn = lift_at(hold_blend, yh, frac)
    rows.append({"calls made (top x%)": f"{frac:.0%}", "recall first": rb, "recall now": rn, "lift first": lb, "lift now": ln,
                 "precision now": ln * yh.mean()})
print("\nCall-budget view:"); print(pd.DataFrame(rows).set_index("calls made (top x%)").round(3).to_string())

fig, axes = plt.subplots(1, 2, figsize=(13, 4.6))
w = .38; xs = np.arange(1, 11)
axes[0].bar(xs - w/2, dt_base["rate %"], w, color="#9e9e9e", label="first notebook"); axes[0].bar(xs + w/2, dt_new["rate %"], w, color="#1b9e77", label="this notebook")
axes[0].axhline(yh.mean() * 100, c="red", ls="--", label="average"); axes[0].set_xlabel("Score decile (1 = highest)"); axes[0].set_ylabel("Conversion rate (%)"); axes[0].legend(); axes[0].set_title("Conversion by score decile (holdout)")
for lab, p, c in [("first notebook", p_base, "#9e9e9e"), ("this notebook", hold_blend, "#1b9e77")]:
    o = np.argsort(-p); cum = np.cumsum(yh[o]) / yh.sum() * 100; axes[1].plot(np.arange(1, len(o) + 1) / len(o) * 100, cum, c=c, label=lab)
axes[1].plot([0, 100], [0, 100], "k--", alpha=.4, label="random"); axes[1].set_xlabel("% of clients called (ranked by score)"); axes[1].set_ylabel("% of subscribers reached"); axes[1].legend(); axes[1].set_title("Cumulative gains (holdout)")
plt.tight_layout(); plt.show()
```

### Findings, assumptions and so what: lift and call budgets
- **Top decile:** the 10% highest-scored clients convert at **56.5%** (lift 4.8×), up from 52.4% (4.5×), and capture **48.3%** of all subscribers, up from 44.8%.
- **At a 20% call budget** the model reaches **65.0%** of subscribers (63% before); at 5% it reaches 30% (27% before) with 70% precision.
- **Gains concentrate at the top of the list.** At a 50% budget the two models are identical (83.5% vs 83.6%). The new model improves **who is called first**, not how many people eventually get called.
- **Reading the chart:** the green bars are higher in decile 1 and lower in deciles 2 to 4, meaning scores separate subscribers more sharply.
- **Assumption:** lift is measured on a random holdout of clients that were *already selected for calling* by the bank, so it describes ranking within that population, not the whole customer base.

## 12. Decision thresholds from economics, chosen **without** looking at the test set

```python
# A call is worth making when  p(subscribe) * VALUE  >  COST  ->  threshold = 1 / (VALUE / COST) on CALIBRATED probabilities.
# The ratio below is an ASSUMPTION to be replaced with the bank's real numbers. Thresholds are learned on training OOF predictions.
rows = []
for ratio in (3, 5, 10, 20):
    thr = 1.0 / ratio
    for lab, pp, idx in [("train (OOF)", oof_cal, train_idx), ("holdout", hold_cal, hold_idx)]:
        yt = y[idx]; call = pp >= thr; conv = yt[call].sum()
        rows.append({"value/cost ratio": ratio, "threshold": round(thr, 3), "set": lab, "calls %": call.mean() * 100,
                     "subscribers reached %": conv / yt.sum() * 100, "precision %": conv / max(call.sum(), 1) * 100,
                     "net value per 1000 clients": (conv * ratio - call.sum()) / len(idx) * 1000})
econ = pd.DataFrame(rows)
print(econ.round(2).to_string(index=False))

# Reference points: call nobody / call everybody
for ratio in (3, 5, 10, 20):
    print(f"ratio {ratio:>2}: calling EVERYONE nets {(yh.mean()*ratio - 1)*1000:8.1f} per 1000 clients (holdout)")

# The first notebook's rule: 'recall >= 80%' picked ON THE TEST SET. Here: choose on train OOF, then read the holdout.
thr80 = float(np.quantile(oof_cal[y[train_idx] == 1], 0.20))             # score that keeps 80% of OOF subscribers
call = hold_cal >= thr80
print(f"\n80%-recall rule (threshold {thr80:.3f} fixed from training OOF): holdout recall = {y[hold_idx][call].sum()/yh.sum():.1%}, "
      f"calls = {call.mean():.1%} of clients, precision = {y[hold_idx][call].mean():.1%}")
# Like-for-like: calls needed for 80% recall if the cut-off is picked on the holdout itself (what the first notebook did)
for lab, p in [("first notebook's model", p_base), ("this notebook's model", hold_blend)]:
    cum = np.cumsum(yh[np.argsort(-p)]) / yh.sum(); k = int(np.searchsorted(cum, 0.80)) + 1
    print(f"  {lab:<24}: {k/len(yh):.1%} of clients must be called to reach 80% recall (cut-off chosen on the holdout itself)")
```

### Findings, assumptions and so what: economics-based thresholds
- **The rule:** call a client when `probability × value > cost`, i.e. when the **calibrated** probability exceeds `cost / value`. This only works with calibrated probabilities, which is why section 8 comes first. The first notebook's thresholds were applied to class-weighted scores and chosen on the test set.
- **The value/cost ratios are assumptions** (3, 5, 10, 20) to be replaced with the bank's real economics. Thresholds are learned on **training out-of-fold predictions** and then simply *read* on the holdout.
- **They transfer:** the share of clients called is within about 0.3 percentage points between training and holdout at every ratio (e.g. ratio 10: 24.3% vs 24.3%; the largest gap is ratio 20: 56.2% vs 56.5%), which confirms the calibration holds on unseen data.

| Value / cost | Threshold | Clients called | Subscribers reached | Net value per 1,000 clients | Call everyone |
|---|---|---|---|---|---|
| 3 | 0.33 | 11.0% | 51.0% | **+70** | −649 |
| 5 | 0.20 | 15.3% | 60.5% | **+200** | −415 |
| 10 | 0.10 | 24.3% | 69.6% | **+571** | +170 |
| 20 | 0.05 | 56.5% | 87.2% | **+1,474** | +1,340 |

- **The model's value is largest when calls are expensive relative to the prize.** At ratios of 3 to 5, calling everyone *loses* money and the model turns a loss into a profit; at 20, almost everyone is worth calling and the model adds about 10%.
- **The first notebook's "80% recall" rule,** compared like-for-like: this model needs **42.6%** of clients called to reach 80% recall, against 43.7% for the first notebook's model (both with the cut-off chosen on the holdout itself). Setting the threshold honestly from training predictions (0.065) gives 79.5% recall on the holdout with 41.7% of clients called. Rules based on recall alone ignore costs.
- **Assumptions:** a conversion is worth the same to the bank whoever the client is, and a call costs the same. Both are simplifications.

## 13. What drives the predictions?

```python
lgb = fitted["LightGBM (tuned)"]
# (a) Permutation importance on the holdout (drop in PR-AUC when a feature is shuffled)
pi = permutation_importance(lgb, X.iloc[hold_idx], yh, scoring="average_precision", n_repeats=5, random_state=RANDOM_SEED, n_jobs=1)
imp = pd.Series(pi.importances_mean, index=X.columns).sort_values(ascending=False)
# (b) Mean |SHAP| from LightGBM's built-in contribution output (no extra package needed)
contrib = lgb.predict(X.iloc[hold_idx], pred_contrib=True)[:, :-1]
shap_imp = pd.Series(np.abs(contrib).mean(axis=0), index=X.columns).sort_values(ascending=False)

fig, axes = plt.subplots(1, 2, figsize=(14, 6.5))
imp.head(15)[::-1].plot.barh(ax=axes[0], color="steelblue"); axes[0].set_title("Permutation importance (drop in PR-AUC)")
shap_imp.head(15)[::-1].plot.barh(ax=axes[1], color="#1b9e77"); axes[1].set_title("Mean |SHAP| (log-odds)")
plt.tight_layout(); plt.show()
print("Top 8 by permutation importance:", [f"{k} ({v:.3f})" for k, v in imp.head(8).items()])
print("Top 8 by mean |SHAP|          :", [f"{k} ({v:.3f})" for k, v in shap_imp.head(8).items()])

# Which signals are era proxies? Share of total SHAP carried by call-plan + calendar features
plan = ["calls_that_day", "calls_that_month", "share_prev_contacted_month", "share_contact_unknown_month"]
cal_f = ["month", "month_num", "day_of_year", "day", "weekday"]
tot = shap_imp.sum()
print(f"\nShare of total |SHAP|:  call-plan {shap_imp[plan].sum()/tot:.1%} | calendar {shap_imp[cal_f].sum()/tot:.1%} | "
      f"client + history {shap_imp.drop(plan + cal_f).sum()/tot:.1%}")
```

### Findings, assumptions and so what: what drives the predictions
- **Two views agree:** permutation importance (drop in PR-AUC when a feature is shuffled) and mean |SHAP| (LightGBM's built-in contributions). **`calls_that_day` is first in both** (permutation 0.098, SHAP 0.363). `contact`, `share_prev_contacted_month`, `calls_that_month` and `prev_success` follow.
- **Where the signal comes from (share of total |SHAP|):** call-plan features **37.5%**, calendar features 16.3%, client and history features **46.3%**.
- **What this means:** a large part of the model's skill comes from the structure of the call schedule. Days with few calls are the warm-list days. That is consistent with the EDA, and it improved results in the forward-in-time tests. It also means the model is **only as portable as the call plan is stable.**
- **Risks to manage:**
  1. The call-plan features must be computed at scoring time in exactly the same way (calls per day and month, list mix). If the dialler schedule changes, retrain.
  2. If the bank now dials in a different regime (different list size, different channel mix), expect the call-plan signal to shift.
  3. `marital` ranks high in SHAP (0.148) despite weak univariate effects in the EDA, which suggests interactions with the schedule or era. This should be investigated and **reviewed for fairness** before any demographic variable is used for targeting.
- **A fallback exists:** feature set B (no call-plan features) still gives CV ROC 0.805 / PR 0.461, already above the first notebook's CV. Use it if the call-plan features cannot be produced reliably.

## 14. Summary, limitations and next steps

### Summary

**What improved (untouched holdout, first notebook's best model → this notebook)**

| Metric | Before | After | Notes |
|---|---|---|---|
| ROC-AUC | 0.806 | **0.816** | +0.010, 95% CI [+0.002, +0.019] |
| PR-AUC | 0.462 | **0.510** | +0.048, CI [+0.033, +0.064], +10% relative |
| Within-era ROC | 0.707 | **0.736** | gains in all three eras |
| Forward-in-time ROC | 0.64 to 0.66 | **0.69 to 0.76** | the number that predicts real deployment |
| Calibration (Brier) | 0.0805 raw | **0.0760** | probabilities now usable for decisions |
| Subscribers reached by calling the top 10% | 44.8% | **48.3%** | |

**What produced the gains, in order of importance**
1. **Seeing the timeline:** reconstructing the dates revealed the 2008 → 2010 regime change and allowed call-plan and weekday features (+0.010 ROC, +0.025 PR in CV, the largest single step).
2. **Not discarding information:** raw `pdays`, weekday and date parts (+0.007 ROC, +0.016 PR in CV).
3. **Honest evaluation:** CV for selection, an untouched holdout, bootstrap CIs, within-era and forward-in-time checks.
4. **Tuning and native categorical handling** (+0.005 ROC in CV). **Blending** added almost nothing, and the notebook says so.
5. **Calibration and economics-based thresholds** that do not use the test set.

**Limits**
- **Ceiling:** without `duration`, expect ROC-AUC of about 0.82 on this dataset. Further gains need new information (client-level history, income, product holdings), not new algorithms.
- **2008 is weakly predictable** (ROC 0.65). Its clients had little history.
- **Observational data from one bank:** associations, not causal effects; no client id, so repeated clients cannot be detected.
- **The call-plan dependence** (section 13) means the model should be re-validated whenever the dialling strategy changes.

**Recommended next steps**
1. Confirm that the call-plan features can be produced at scoring time. If not, use feature set B.
2. Replace the assumed value/cost ratios (section 12) with real economics.
3. Run a **randomised pilot**: score clients, call the top 20% versus a random 20%, and measure the real lift.
4. Monitor monthly: PR-AUC, calibration (predicted vs actual rate) and call-plan feature drift. Retrain on a schedule.
5. Have compliance review the use of age, marital status and similar attributes for targeting.
