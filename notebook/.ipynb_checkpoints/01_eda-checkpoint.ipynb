# Bank Marketing Campaign
---
**Dataset:** [UCI Bank Marketing Dataset](https://archive.ics.uci.edu/dataset/222/bank+marketing)  
**Business Problem:** A Portuguese bank ran direct phone-call marketing campaigns to sell term deposits.  
The goal of this EDA is to understand *which customer segments convert*, *when to call*, and *how prior contact history affects outcomes* — so that future campaigns can be better targeted.

## Data dictionary
| Feature | Description | Type |
|---|---|---|
| `age` | Age in years | Numeric |
| `job` | Type of job (admin., blue-collar, technician, management, services, retired, self-employed, entrepreneur, unemployed, housemaid, student, unknown) | Categorical |
| `marital` | Marital status (married, single, divorced). "Divorced" includes widowed | Categorical |
| `education` | Education level (primary, secondary, tertiary, unknown) | Categorical |
| `default` | Has credit in default? (yes/no) | Binary |
| `balance` | Average yearly account balance (EUR) | Numeric |
| `housing` | Has a housing loan? (yes/no) | Binary |
| `loan` | Has a personal loan? (yes/no) | Binary |
| `contact` | Contact communication type (cellular, telephone, unknown) | Categorical |
| `day` | Day of the month of the last contact (1 to 31) | Numeric |
| `month` | Month of the last contact (jan to dec). **No year is provided** | Categorical |
| `duration` | Duration of the last call in seconds. **Leakage: known only after the call** | Numeric |
| `campaign` | Number of contacts made to this client during the current campaign (includes the last contact) | Numeric |
| `pdays` | Days since the client was last contacted in a previous campaign (`-1` = never contacted, a **sentinel**, not a number) | Numeric |
| `previous` | Number of contacts made to this client before the current campaign | Numeric |
| `poutcome` | Outcome of the previous marketing campaign (success, failure, other, unknown) | Categorical |
| `y` | **Target.** Did the client subscribe to a term deposit? (yes/no) | Binary |

## 1. Environment & Imports <a id='1'></a>

```python
import pandas as pd
import numpy as np
import matplotlib.ticker as mtick
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import chi2_contingency, mannwhitneyu, ttest_ind, shapiro, probplot, spearmanr
from statsmodels.stats.proportion import proportions_ztest, proportion_confint
import warnings
warnings.filterwarnings('ignore')

RANDOM_SEED = 42
np.random.seed(RANDOM_SEED)


sns.set_theme(style='whitegrid', palette='muted')
plt.rcParams.update({
    'figure.dpi': 120,
    'figure.figsize': (6, 4),
    'axes.titlesize': 14,
    'axes.labelsize': 12,
    'xtick.labelsize': 10,
    'ytick.labelsize': 10,
})
test_log = []
ALPHA = 0.05
```

```python
# Helper functions so we do not repeat the same code again and again

# Conversion rate (%) for every category of a column, with 95% Wilson confidence interval.
def conv_table(data, col):
    t = data.groupby(col, observed=True)['y_num'].agg(total='count', converted='sum').reset_index()
    t['conv_rate'] = t['converted'] / t['total'] * 100
    ci = [proportion_confint(c, n, alpha=ALPHA, method='wilson') for c, n in zip(t['converted'], t['total'])]
    t['ci_low'] = [c[0] * 100 for c in ci]
    t['ci_high'] = [c[1] * 100 for c in ci]
    return t

# Chi-square test of independence between a categorical column and the target.
# Also returns Cramer's V (effect size) and the smallest expected count (assumption check).
def chi_square_test(data, col, label=None):
    table = pd.crosstab(data[col], data['y'])
    chi2, p, dof, expected = chi2_contingency(table)
    cramers_v = np.sqrt(chi2 / (table.values.sum() * (min(table.shape) - 1)))
    test_log.append({'feature': label or col, 'test': 'Chi-square', 'statistic': chi2,
                     'p_value': p, 'effect_size': cramers_v, 'effect_name': "Cramer's V"})
    print(f"Chi-square test ({col} vs y): chi2 = {chi2:.2f}, dof = {dof}, p = {p:.2e}")
    print(f"Cramer's V (effect size) = {cramers_v:.3f}")
    print(f"Smallest expected count  = {expected.min():.1f}  (should be >= 5 for the test to be valid)")

# Compare a numeric column between y = yes and y = no.
# Mann-Whitney U is the main test (no normality needed). Welch t-test is shown for comparison.
def numeric_vs_target(data, col):
    
    yes = data.loc[data['y'] == 'yes', col]
    no = data.loc[data['y'] == 'no', col]
    u, p_mw = mannwhitneyu(yes, no, alternative='two-sided')
    t, p_t = ttest_ind(yes, no, equal_var=False)
    # rank-biserial correlation: effect size for Mann-Whitney (0 = no effect, +-1 = complete separation)
    rbc = 2 * u / (len(yes) * len(no)) - 1
    print(f"{col}: median yes = {yes.median():.0f}, median no = {no.median():.0f} | mean yes = {yes.mean():.1f}, mean no = {no.mean():.1f}")
    print(f"Mann-Whitney U : p = {p_mw:.2e}, rank-biserial r = {rbc:.3f}")
    print(f"Welch t-test   : t = {t:.2f}, p = {p_t:.2e}")
    test_log.append({'feature': col, 'test': 'Mann-Whitney U', 'statistic': u,
                     'p_value': p_mw, 'effect_size': rbc, 'effect_name': 'rank-biserial r'})


# Two-proportion z-test: is the conversion rate of group_a different from group_b?
def two_group_ztest(data, col, group_a, group_b):

    a = data[data[col] == group_a]['y_num']
    b = data[data[col] == group_b]['y_num']
    z, p = proportions_ztest([a.sum(), b.sum()], [len(a), len(b)])
    print(f"{col}: {group_a} = {a.mean()*100:.1f}% (n={len(a):,})  vs  {group_b} = {b.mean()*100:.1f}% (n={len(b):,})  ->  z = {z:.2f}, p = {p:.2e}")

# Bar chart of conversion rate with 95% CI error bars and a dashed baseline.
def bar_with_ci(ax, t, xcol, horizontal=False, baseline=None, title=''):
    err = [t['conv_rate'] - t['ci_low'], t['ci_high'] - t['conv_rate']]
    if horizontal:
        ax.barh(t[xcol].astype(str), t['conv_rate'], xerr=err, color='steelblue', edgecolor='white', capsize=3)
        if baseline is not None:
            ax.axvline(baseline, color='red', linestyle='--', label=f'Baseline {baseline:.1f}%')
        ax.set_xlabel('Conversion Rate (%)')
    else:
        ax.bar(t[xcol].astype(str), t['conv_rate'], yerr=err, color='steelblue', edgecolor='white', capsize=3)
        if baseline is not None:
            ax.axhline(baseline, color='red', linestyle='--', label=f'Baseline {baseline:.1f}%')
        ax.set_ylabel('Conversion Rate (%)')
    ax.set_title(title)
    ax.legend()
```

## 3. Data Quality Audit <a id='3'></a>
We check for: nulls, 'unknown' placeholders, impossible values, and encoding quirks.


```python
# checking for null values in the dataset
df.isnull().sum()

# No null values found
```

```python
# 3b. 'unknown' placeholders in categorical columns 
cat_cols = df.select_dtypes(include='object').columns.tolist()

unknown_report = []
for col in cat_cols:
    n = (df[col] == 'unknown').sum()
    pct = n / len(df) * 100
    if n > 0:
        unknown_report.append({'column': col, 'unknown_count': n, 'unknown_pct': round(pct, 2)})

unknown_df = pd.DataFrame(unknown_report).sort_values('unknown_pct', ascending=False)
print("Columns with 'unknown' values:")
print(unknown_df.to_string(index=False))
```

**Key findings:**
* `poutcome` 81.8% unknown, `contact` 28.8%, `education` 4.1%, `job` 0.6%.
* `poutcome = unknown` is not really "missing" - it means the client was **never contacted before** (checked in the next cell).

**Decision:**
* Keep `unknown` as its own category. Dropping rows would remove 82% of the data (poutcome) and the unknown group may behave differently - we will see later that unknown `contact` converts at only 4%.

```python
# # 3c. pdays special encoding: -1 means 'never previously contacted'
# pdays_neg = (df['pdays'] == -1).sum()
# print(f"pdays = -1 (never contacted): {pdays_neg:,} rows  ({pdays_neg/len(df)*100:.1f}%)")
# print(f"pdays > 0  (was contacted)  : {(df['pdays'] > 0).sum():,} rows")

# # Create a cleaner binary flag and a filtered series for real values
# df['was_previously_contacted'] = (df['pdays'] != -1).astype(int)
# pdays_real = df.loc[df['pdays'] != -1, 'pdays']
# print(f"\npdays stats (excluding -1 sentinel):")
# print(pdays_real.describe().round(1))

# 3c. pdays = -1 (client was never contacted before)
never_contacted = (df['pdays'] == -1).sum()
print(f"pdays = -1 (never contacted before): {never_contacted:,} rows ({never_contacted/len(df)*100:.1f}%)")

# Consistency check: pdays and previous should tell the same thing
print("Rows with pdays = -1 but previous > 0 :", ((df['pdays'] == -1) & (df['previous'] > 0)).sum())
print("Rows with pdays > 0  but previous = 0 :", ((df['pdays'] > 0) & (df['previous'] == 0)).sum())
print("Rows with poutcome = 'unknown' but client WAS contacted before:",
      ((df['poutcome'] == 'unknown') & (df['pdays'] != -1)).sum())

# New binary column: was the client contacted in an earlier campaign?
df['was_previously_contacted'] = (df['pdays'] != -1).astype(int)

# pdays statistics only for clients who were really contacted before
print("\npdays summary (excluding -1):")
print(df.loc[df['pdays'] != -1, 'pdays'].describe().round(1))
```

**Key findings:**
* 36,954 clients (81.7%) were never contacted before (*pdays = -1*). *pdays* and *previous* are consistent (no contradictions).
* Only **5 rows** have *poutcome = unknown* although they were contacted before - negligible.
* Among the 8,257 previously contacted clients, the median gap since the last contact is 194 days.

**Decision:** **-1** is a code, not a number of days. So we create *was_previously_contacted* (0/1) and analyse real *pdays* values separately.

```python
# 3d. Check numeric ranges for impossible values
print(df[num_cols].describe().T[['min', 'max', 'mean', '50%']].round(2))

print("\nCalls with duration = 0 seconds:", (df['duration'] == 0).sum())
print("Clients with negative balance  :", (df['balance'] < 0).sum())
print("Largest value of 'previous'     :", df['previous'].max(), "99th percentile is", round(df['previous'].quantile(0.99), 1))
```

**Key findings:**
* Ages 18-95: realistic.
* 3,766 clients (8.3%) have a negative balance - valid.
* Max **balance** = 102,127 EUR vs median 448 EUR - very large values, but possible for rich customers, so **we keep them**.

* *previous* has a maximum of 275 contacts while the 99th percentile is only ~9. That single extreme looks like a data-entry error.

## 4. Class Imbalance (Target Distribution) <a id='4'></a>
Every acceptance rate computed below must be interpreted relative to the **baseline conversion rate**.


```python
target_counts = df['y'].value_counts()
target_pct    = df['y'].value_counts(normalize=True) * 100

BASELINE_CONVERSION = target_pct['yes']

print("=== Target Distribution ===")
for label in ['yes', 'no']:
    print(f"  {label:>3}  {target_counts[label]:>6,}  ({target_pct[label]:.1f}%)")
print(f"\nBaseline conversion rate: {BASELINE_CONVERSION:.2f}%")
print("Class imbalance ratio (No:Yes):", round(target_counts['no'] / target_counts['yes'], 1), ":1")
```

**Key findings:**
* Baseline conversion = **11.70%** (95% CI 11.41% - 12.00%): about 1 in 9 calls succeeds.
* Imbalance is 7.5 : 1 (No : Yes).
* A "dumb" model that always says *no* would be **88.3% accurate** while finding zero buyers.

**Implication:** accuracy is a misleading metric. For modelling use ROC-AUC, PR-AUC or F1 on the *yes* class, with stratified train/test split and class weights.

**Assumption:** the dataset represents calls in general. If the bank only called "promising" clients, 11.7% would be higher than the true conversion of the whole customer base.

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 4))

# Bar chart
colors = ['#e74c3c', '#2ecc71']
axes[0].bar(target_counts.index, target_counts.values, color=colors, edgecolor='white', linewidth=1.5)
axes[0].set_title('Target Class Distribution')
axes[0].set_ylabel('Number of Clients')
axes[0].set_xlabel('Subscribed Term Deposit (y)')

# Pie chart
axes[1].pie(target_counts.values, labels=target_counts.index,
            autopct='%1.1f%%', colors=colors, startangle=90,
            wedgeprops={'edgecolor': 'white', 'linewidth': 2})
axes[1].set_title('Class Split')

plt.suptitle('Class Imbalance: ~88% No vs ~12% Yes', fontsize=13, y=1.02)
plt.tight_layout()
plt.show()
```

## 5. Univariate Analysis — Numerical Features <a id='5'></a>

### 5.1 Age

```python
print(df['age'].describe().round(2))
print(f"Skewness : {df['age'].skew():.3f}")
print(f"Kurtosis : {df['age'].kurtosis():.3f}")

# Shapiro-Wilk normality test (on a random sample of 5000, because the test is very sensitive for large data)
stat, p = shapiro(df['age'].sample(5000, random_state=RANDOM_SEED))
print(f"\nShapiro-Wilk on 5000 sampled ages: W = {stat:.3f}, p = {p:.2e}")

# Outliers using the IQR rule
Q1, Q3 = df['age'].quantile([0.25, 0.75])
IQR = Q3 - Q1
lower, upper = Q1 - 1.5 * IQR, Q3 + 1.5 * IQR
n_out = ((df['age'] < lower) | (df['age'] > upper)).sum()
print(f"IQR fence: [{lower:.0f}, {upper:.0f}] | Outliers: {n_out:,} ({n_out/len(df)*100:.1f}%)")
```

**Key findings:**
* Mean 40.9, median 39, range 18-95. Skewness 0.69 (moderate right skew), kurtosis 0.32.
* **Shapiro-Wilk p ≈ 2e-34 -> age is NOT perfectly normal.** So for age we prefer **rank-based tests**
* 487 clients (1.1%) are above the IQR fence of 70 years - these are real elderly clients, not errors.

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

# Histogram + KDE
df['age'].plot(kind='hist', bins=30, density=True, alpha=0.5, ax=axes[0], label='Histogram', color='steelblue')
df['age'].plot(kind='kde', ax=axes[0], label='KDE', color='navy')
axes[0].set_title('Age Distribution')
axes[0].set_xlabel('Age')
axes[0].legend()

# Boxplot
sns.boxplot(y=df['age'], ax=axes[1], color='steelblue')
axes[1].set_title('Age Boxplot')

# Q-Q plot
probplot(df['age'], plot=axes[2])
axes[2].set_title('Age Q-Q Plot')

plt.tight_layout()
plt.show()
# The Age column is roughly normally distributed. Though it seems there are some outliers but they are in range of 100 so it can be actual values.
```

The histogram peaks around age 32, falls sharply after ~60 (probably retirement age) and has a thin tail up to 95. The Q-Q plot leaves the red line at both ends: the lower end is bounded at 18 and the upper end rises faster than a normal curve (heavier right tail), so it is not normal, which agrees with Shapiro-Wilk. The boxplot shows clients above 70 as separate dots - expected for a bank's customers.

## 7. Bivariate Analysis — Feature vs Target <a id='7'></a>
Each feature is compared with the target in the same 3 steps:   
**(1) conversion table + CI,   
(2) hypothesis test + effect size,   
(3) chart with the baseline (red dashed line)**.

For each test we state the hypotheses:
* **H₀ (null):** the feature and `y` are independent (conversion is the same in every group).
* **H₁ (alternative):** conversion differs between groups.
* Decision rule: reject H₀ if p < 0.05


### 7.1 Age vs Target

```python
# Age bins for cleaner business interpretation
df['age_band'] = pd.cut(df['age'], bins=[17,25,35,45,55,65,100],
                         labels=['18-25','26-35','36-45','46-55','56-65','65+'])

age_acc = (
    df.groupby('age_band', observed=True)
    .agg(total=('y','count'), converted=('y', lambda x: (x=='yes').sum()))
    .assign(conv_rate=lambda x: x['converted']/x['total']*100)
    .reset_index()
)

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

axes[0].bar(age_acc['age_band'].astype(str), age_acc['conv_rate'], color='steelblue', edgecolor='white')
axes[0].axhline(BASELINE_CONVERSION, color='red', linestyle='--', linewidth=1.5, label=f'Baseline {BASELINE_CONVERSION:.1f}%')
axes[0].set_title('Conversion Rate by Age Band')
axes[0].set_xlabel('Age Band')
axes[0].set_ylabel('Conversion Rate (%)')
axes[0].legend()
for i, row in age_acc.iterrows():
    axes[0].text(i, row['conv_rate']+0.3, f"{row['conv_rate']:.1f}%", ha='center', fontsize=9)

sns.boxplot(x='y', y='age', data=df, palette={'yes':'#2ecc71','no':'#e74c3c'}, ax=axes[1])
axes[1].set_title('Age Distribution by Subscription')
axes[1].set_xlabel('Subscribed')
axes[1].set_ylabel('Age')

plt.tight_layout()
plt.show()

# Statistical test
# Chi-Square Test
ct = pd.crosstab(df['age_band'], df['y'])
chi2, p, dof, expected = chi2_contingency(ct)
print(f"\nChi-square test: chi2={chi2:.2f}, p={p:.2e}, dof={dof}")

# Cramer-V Test
n = ct.to_numpy().sum()
r, k = ct.shape
cramers_v = np.sqrt(chi2 / (n * min(r - 1, k - 1)))
print(cramers_v)
print("Minimum expected frequency:", round(expected.min(),2))
```

**Key findings:**
* Conversion by age band is **U-shaped**: 24.0% (18-25), 12.0%, 9.4%, 9.4%, 14.1% (56-65), and **42.6% for 66+** (n = 751; CI 39.1% - 46.2%).
* Chi-square: χ² = 1036, dof = 5, p ≈ 1e-221 -> reject H₀; age band and subscription are associated. Cramér's V = 0.15 (small-to-medium effect).
* Smallest expected count = 87.9 (≥ 5), so the chi-square assumption is satisfied.

**Insight:** the youngest (<=25) and oldest (66+) clients are the best segments. People between 36 and 55 are *below* the 11.7% baseline.

**Assumption:** rows are independent. Age bands were chosen by us (not by the data), so the exact cut points are a judgement.

### 7.2 Job vs Target

```python
# Statistical test
# Chi-Square Test
ct = pd.crosstab(df['job'], df['y'])
chi2, p, dof, expected = chi2_contingency(ct)
print(f"\nChi-square test: chi2={chi2:.2f}, p={p:.2e}, dof={dof}")

# Cramer-V Test
n = ct.to_numpy().sum()
r, k = ct.shape
cramers_v = np.sqrt(chi2 / (n * min(r - 1, k - 1)))
print(cramers_v)
print("Minimum expected frequency:", round(expected.min(),2))

job_acc = (
    df.groupby('job')
    .agg(total=('y','count'), converted=('y', lambda x: (x=='yes').sum()))
    .assign(conv_rate=lambda x: x['converted']/x['total']*100)
    .sort_values('conv_rate', ascending=True)
    .reset_index()
)

fig, ax = plt.subplots()
bars = ax.barh(job_acc['job'], job_acc['conv_rate'],
               color=['#2ecc71' if v > BASELINE_CONVERSION else '#e74c3c' for v in job_acc['conv_rate']])
ax.axvline(BASELINE_CONVERSION, color='navy', linestyle='--', linewidth=1.5, label=f'Baseline {BASELINE_CONVERSION:.1f}%')
ax.set_title('Conversion Rate by Job (green = above baseline)')
ax.set_xlabel('Conversion Rate (%)')
ax.legend()
for i, row in job_acc.iterrows():
    ax.text(row['conv_rate']+0.1, i, f"{row['conv_rate']:.1f}%", va='center', fontsize=9)
plt.tight_layout()
plt.show()
```

**Key findings:**
* Best: **student 28.7%** (CI 25.9-31.7%) and **retired 22.8%** (CI 21.1-24.6%) - roughly 2-2.5× the baseline.
* Worst: blue-collar 7.3%, entrepreneur 8.3%, housemaid 8.8%, services 8.9% - all below the baseline.
* Chi-square: χ² = 836, dof = 11, p ≈ 3e-172 -> job and `y` are associated. Cramér's V = 0.14.
* Smallest expected count = 33.7 -> assumption met.

**Important:**
* Student and retired are exactly the groups at the two ends of the age U-shape. Part of the "job effect" may really be an **age effect**. 

### 7.3 Education vs Target

```python
edu_acc = (
    df.groupby('education')
    .agg(total=('y','count'), converted=('y', lambda x: (x=='yes').sum()))
    .assign(conv_rate=lambda x: x['converted']/x['total']*100)
    .sort_values('conv_rate', ascending=False)
    .reset_index()
)
print(edu_acc.to_string(index=False))

# Chi-square test
# Hypotheses. 
# H₀: education and y are independent (subscription rate is the same across education levels). 
# H₁: they are associated.
ct = pd.crosstab(df['education'], df['y'])
chi2, p, dof, _ = chi2_contingency(ct)
print(f"\nChi-square test: chi2={chi2:.2f}, p={p:.2e}, dof={dof}")
```

**Key findings:**
* Conversion rises with education: primary 8.6% -> secondary 10.6% -> tertiary 15.0%. **unknown** (13.6%) sits close to tertiary.
* Chi-square: χ² = 238.9, dof = 3, p = 1.6e-51 -> reject H₀. **Cramér's V = 0.073 - small effect** (very significant p-value, but education explains little on its own).

### 7.4 Marital Status vs Target

```python
# Conversion rate
mar_acc = (
    df.groupby('marital')
    .agg(total=('y','count'), converted=('y', lambda x: (x=='yes').sum()))
    .assign(conv_rate=lambda x: x['converted']/x['total']*100)
    .reset_index()
)
print(mar_acc.to_string(index=False))


plt.bar(mar_acc['marital'], mar_acc['conv_rate'], color=sns.color_palette('muted',3), edgecolor='white')
plt.axhline(BASELINE_CONVERSION, color='red', linestyle='--', label=f'Baseline')
plt.title('Conversion Rate by Marital Status')
plt.ylabel('Conversion Rate (%)')
plt.legend()
plt.show()

#Statistics
ct = pd.crosstab(df['marital'], df['y'])
chi2, p, dof, _ = chi2_contingency(ct)
print(f"\nChi-square test: chi2={chi2:.2f}, p={p:.2e}, dof={dof}")
cramers_v = np.sqrt(chi2 / (n * min(r - 1, k - 1)))
print(f"Cramer's V (effect size) {round(cramers_v,2)}")
print("Minimum expected frequency:", round(expected.min(),2))
```

**Key findings:**
* Single 14.9% > divorced 11.9% > married 10.1%.
* Chi-square: χ² = 196.5, dof = 2, p = 2e-43; Cramér's V = 0.07 (small), So we reject NULL Hypothesis.

**Important:** marital status is related to age (singles are younger). Some of this effect may come from age.

### 7.5 Balance vs Target

```python
# Numeric test first
numeric_vs_target(df, 'balance')

# Then balance quartiles: gives a business-friendly view of the same relationship
df['balance_quartile'] = pd.qcut(df['balance'], 4, labels=['Q1 (lowest)', 'Q2', 'Q3', 'Q4 (highest)'])
bal_acc = conv_table(df, 'balance_quartile')
print()
print(bal_acc.round(1).to_string(index=False))
print()
chi_square_test(df, 'balance_quartile', label='balance (in quartiles)')

fig, axes = plt.subplots(1, 2, figsize=(13, 5))
clip_val = df['balance'].quantile(0.99)
sns.boxplot(x='y', y='balance', hue='y', data=df[df['balance'] < clip_val],
            palette={'yes': '#2ecc71', 'no': '#e74c3c'}, legend=False, ax=axes[0])
axes[0].set_title('Balance by Subscription (below 99th pct)')
bar_with_ci(axes[1], bal_acc, 'balance_quartile', baseline=BASELINE_CONVERSION, title='Conversion Rate by Balance Quartile')
plt.tight_layout()
plt.show()
```

**Key findings:**
* Clients who subscribe have higher balances: median 733 vs 417 EUR (mean 1,804 vs 1,304).
* **Mann-Whitney p ≈ 7e-101, rank-biserial r = 0.18** (small-to-medium effect). Welch t-test agrees (p ≈ 4e-23).
* Quartile view: conversion rises steadily from **7.2% (lowest 25% balances) to 16.2% (highest 25%)** - a clear monotonic trend. Chi-square on quartiles: p ≈ 4e-97, V = 0.10.

**Why Mann-Whitney and not t-test:** balance has skewness 8.4 and extreme outliers. The t-test compares means, which a few huge balances can pull strongly; Mann-Whitney compares ranks and is not affected. Both agree here, so the conclusion is robust.

**Assumption:** the two groups are independent samples. The quartile bins are equal-sized by construction, so no group is small.

### 7.6 Housing & Personal Loan vs Target

**Key findings:**

* **Personal loan:** no 12.7% vs yes 6.7% (p ≈ 2e-47, V = 0.07).
* **Default:** no 11.8% vs yes 6.4% (p ≈ 2e-6, V = 0.02 - negligible; only 815 clients, CI 4.9-8.3%).

**Interpretation (a hypothesis, not proven):** clients who already have loan payments have less free cash to put into a term deposit.

### 7.7 Contact Type vs Target

```python
cont_acc = conv_table(df, 'contact')
print(cont_acc.round(1).to_string(index=False))
print()
chi_square_test(df, 'contact')
print()
two_group_ztest(df, 'contact', 'cellular', 'telephone')
two_group_ztest(df, 'contact', 'cellular', 'unknown')

# Where do the 'unknown' contacts come from? (check month)
print("\nShare of 'unknown' contact type by month (%):")
print((pd.crosstab(df['month'], df['contact'], normalize='index')['unknown'] * 100).round(0).sort_values(ascending=False).head(4))

fig, ax = plt.subplots(figsize=(7, 4))
bar_with_ci(ax, cont_acc, 'contact', baseline=BASELINE_CONVERSION, title='Conversion Rate by Contact Type (with 95% CI)')
plt.tight_layout()
plt.show()
```

**Key findings:**

* Chi-square: p ≈ 1e-225, V = 0.15. But the effect comes almost entirely from **unknown**, not from cellular-vs-telephone.
* **Cellular vs telephone: z = 2.17, p = 0.030** - significant at 5% but **not** after Bonferroni correction (alpha ≈ 0.003) and only 1.5 percentage points apart.
* **unknown** contact is concentrated in **June (85%) and May (58%)**.

### 7.8 Month vs Target

```python
month_order = ['jan', 'feb', 'mar', 'apr', 'may', 'jun', 'jul', 'aug', 'sep', 'oct', 'nov', 'dec']

monthly_acc = conv_table(df, 'month')
monthly_acc['month'] = pd.Categorical(monthly_acc['month'], categories=month_order, ordered=True)
monthly_acc = monthly_acc.sort_values('month').reset_index(drop=True)
print(monthly_acc.round(1).to_string(index=False))
print()
chi_square_test(df, 'month')

fig, axes = plt.subplots(2, 1, figsize=(8, 6))

axes[0].errorbar(monthly_acc['month'].astype(str), monthly_acc['conv_rate'],
                 yerr=[monthly_acc['conv_rate'] - monthly_acc['ci_low'], monthly_acc['ci_high'] - monthly_acc['conv_rate']],
                 marker='o', color='steelblue', linewidth=2, capsize=4)
axes[0].axhline(BASELINE_CONVERSION, color='red', linestyle='--', label=f'Baseline {BASELINE_CONVERSION:.1f}%')
axes[0].set_title('Conversion Rate by Month (with 95% CI)')
axes[0].set_ylabel('Conversion Rate (%)')
axes[0].legend()

axes[1].bar(monthly_acc['month'].astype(str), monthly_acc['total'], color='lightsteelblue', edgecolor='white')
axes[1].set_title('Call Volume by Month')
axes[1].set_ylabel('Number of Calls')

plt.tight_layout()
plt.show()
```

**Key findings:**
* Highest rates: **March 52.0%, December 46.7%, September 46.5%, October 43.8%**. Next: April 19.7%, Feb 16.6%.
* Lowest: **May 6.7%** - but May has the most calls (13,766, 30% of all calls).
* Chi-square: χ² = 3,062, dof = 11, p ≈ 0; **V = 0.26** - month has the second largest effect among categorical features.

**Important:** the four "best" months have **very few calls** (214 to 738 each; together 2,008 = 4.4% of all calls) and wide confidence intervals (e.g. December 40.2-53.4%). A high rate with few calls usually means the bank called a **hand-picked audience** in those months (e.g. clients with prior success). Calling more people in March will probably **not** keep a 52% rate.

## 8. Multivariate Analysis <a id='8'></a>
So far each feature was studied alone. Here we check whether the effects **survive when features are considered together**.

### 8.1 Correlation Heatmap - Numerical Features

```python
# Spearman correlation is used because balance, duration, campaign, pdays and previous are highly skewed
corr_cols = ['age', 'balance', 'day', 'duration', 'campaign', 'pdays', 'previous', 'y_num']
corr = df[corr_cols].corr(method='spearman')

fig, ax = plt.subplots(figsize=(6, 4))
mask = np.triu(np.ones_like(corr, dtype=bool))
sns.heatmap(corr, mask=mask, annot=True, fmt='.2f', cmap='coolwarm', center=0, linewidths=0.5, ax=ax)
ax.set_title('Spearman Correlation Matrix')
plt.tight_layout()
plt.show()

# Correlation of each feature with the target, with p-values
print("Spearman correlation with y (1 = yes):")
for col in ['age', 'balance', 'day', 'duration', 'campaign', 'pdays', 'previous']:
    r, p = spearmanr(df[col], df['y_num'])
    print(f"  {col:<9} r = {r:+.3f}   p = {p:.2e}")
```

**Key findings (Spearman):**
* With the target: **duration** +0.34 (leakage, ignore), **previous** +0.17, **pdays** +0.15, **balance** +0.10, **campaign** -0.08, **day** -0.03, **age** -0.01 (not significant, p = 0.063).
* All correlations with **y** are weak (< 0.2) apart from the leaking **duration**. This is expected for a binary target and non-linear effects (e.g. the U-shape of age).
* **pdays** and **previous** are almost the same variable (Spearman ≈ 0.99, both are "not contacted" for the same 82% of clients) - **do not put both in a linear model** (multicollinearity).

**Assumption:** Spearman measures monotonic relationships only. Age shows ~0 correlation even though it has a strong U-shaped effect - a reminder that "no correlation" does not mean "no relationship".

### 8.2 Golden Segments — Job × Education

```python
seg = df.groupby(['job', 'education'])['y_num'].agg(total='count', conv_rate='mean').reset_index()
seg['conv_rate'] = seg['conv_rate'] * 100

# Hide combinations with fewer than 30 clients - their rates are too unstable to trust
MIN_SIZE = 30
print("Job x education combinations with fewer than", MIN_SIZE, "clients:", (seg['total'] < MIN_SIZE).sum(), "of", len(seg))
seg.loc[seg['total'] < MIN_SIZE, 'conv_rate'] = np.nan

pivot = seg.pivot(index='job', columns='education', values='conv_rate').round(1)

fig, ax = plt.subplots(figsize=(6, 4))
sns.heatmap(pivot, annot=True, fmt='.1f', cmap='YlGn', linewidths=0.5, ax=ax, cbar_kws={'label': 'Conversion Rate (%)'})
ax.set_title('Conversion Rate (%) - Job x Education (blank = fewer than 30 clients)')
plt.tight_layout()
plt.show()

print("Top 5 segments (with at least 100 clients):")
print(seg[seg['total'] >= 100].sort_values('conv_rate', ascending=False).head(5).round(1).to_string(index=False))
```

**Key findings:**

* Among groups with at least 100 clients, the best are **students (secondary 29.7%, tertiary 26.5%)** and **retired (tertiary 27.6%)**.
* **Education matters inside most jobs:** for admin., blue-collar, self-employed and technician, tertiary clients convert about 6-12 points higher than primary clients (e.g. blue-collar 5.8% -> 16.1%).
* **Student and retired are high at every education level** (21-36%). For students the pattern even reverses (primary 36.4% vs tertiary 26.5%), but these cells are small, so do not over-read it.

**Interpretation:** both job and education carry information. The student/retired effect is strong at all education levels, which fits the age explanation found in 8.4.

### 8.3 Balance vs Age, coloured by Target

Interest is slightly more common at the extreme ages and at higher balances, but **no two-variable rule separates the groups**.

**Conclusion:** the signal is weak in any single pair of numeric features, so a multivariate model (next) is more useful than looking for a visual rule.

## 9. Campaign & Contact History <a id='9'></a>

```python
df['campaign_capped'] = df['campaign'].clip(upper=10)
camp_acc = conv_table(df, 'campaign_capped')
print(camp_acc.round(1).to_string(index=False))
print()
chi_square_test(df, 'campaign_capped', label='campaign (capped at 10)')

# The chi-square test only says "different". To test the DIRECTION (more calls -> lower conversion) we use Spearman
r, p = spearmanr(df['campaign'], df['y_num'])
print(f"Spearman (campaign vs y): r = {r:.3f}, p = {p:.2e}")

fig, ax = plt.subplots(figsize=(6, 4))
ax.errorbar(camp_acc['campaign_capped'], camp_acc['conv_rate'],
            yerr=[camp_acc['conv_rate'] - camp_acc['ci_low'], camp_acc['ci_high'] - camp_acc['conv_rate']],
            marker='o', color='steelblue', linewidth=2, capsize=4)
ax.axhline(BASELINE_CONVERSION, color='red', linestyle='--', label=f'Baseline {BASELINE_CONVERSION:.1f}%')
ax.set_title('Conversion Rate vs Number of Contacts (10 = 10 or more)')
ax.set_xlabel('Contacts in this campaign')
ax.set_ylabel('Conversion Rate (%)')
ax.legend()
plt.tight_layout()
plt.show()
```

**Key findings:**
* Conversion falls from **14.6% (1 call)** to 11.2% (2 and 3 calls) and then keeps falling: 9.0% (4), 7.9% (5), 7.1% (6) ... **4.2% (10 or more)**.
* Chi-square p ≈ 9e-70 (V = 0.09). Spearman r = -0.084, p ≈ 1e-71: more calls goes with lower conversion.

**Important:** clients who say *yes* usually stop being called, while clients who say *no* keep receiving calls. So high call counts are naturally filled with "no" clients. This does **not** prove that extra calls reduce interest - only that persistent callers rarely convert.

**Business suggestion:** a cap of about 3-5 contacts looks reasonable (rates are below 8% beyond 5 calls and 4% beyond 10).

```python
# Days since last contact, only for clients who were contacted before
old = df[df['pdays'] != -1].copy()
old['pdays_band'] = pd.cut(old['pdays'], bins=[0, 90, 180, 365, 900], labels=['1-90', '91-180', '181-365', '366+'])
pdays_acc = conv_table(old, 'pdays_band')
print(pdays_acc.round(1).to_string(index=False))
print()
chi_square_test(old, 'pdays_band', label='pdays (bands, previously contacted only)')

fig, axes = plt.subplots(1, 2, figsize=(13, 4))
for label, color in [('no', '#e74c3c'), ('yes', '#2ecc71')]:
    axes[0].hist(old.loc[old['y'] == label, 'pdays'].clip(upper=400), bins=40, alpha=0.5, color=color, label=label, density=True)
axes[0].set_title('Days Since Last Contact (clipped at 400)')
axes[0].set_xlabel('pdays')
axes[0].legend(title='Subscribed')

bar_with_ci(axes[1], pdays_acc, 'pdays_band', baseline=BASELINE_CONVERSION, title='Conversion by Days Since Last Contact')
plt.tight_layout()
plt.show()
```

**Key findings (previously contacted clients only):**
* 1-90 days since last contact: 35.0%; 91-180: 27.4%; 181-365: 17.8%; 366+: 29.2%.
* Chi-square p ≈ 7e-36, V = 0.14.
* Recent contact is better, **but the pattern is not a straight line**: clients last contacted more than a year ago convert better than those contacted 6-12 months ago.

**Interpretation:** we cannot tell why. It may be a real "re-engagement" effect or a different mix of clients in the old campaign. The 366+ group is small (643). Do not turn this into a simple "call within 90 days" rule without more evidence.

## 10. Key Findings Summary <a id='10'></a>

| # | Feature | Finding | Business Action |
|---|---------|---------|-----------------|
| 1 | **Class Imbalance** | ~88% No, ~12% Yes. Severe imbalance. | Use AUC-ROC / F1 for modelling; consider SMOTE or class weights. |
| 2 | **poutcome = success** | ~65% conversion vs 11% baseline — 6× lift | Prioritise clients with previous successful campaign contact. |
| 3 | **Age 65+** | Highest conversion rate among age bands | Include in premium targeting segment. |
| 4 | **Job: student, retired** | Both above baseline | Target younger students and retirees. |
| 5 | **Month: Mar, Sep, Oct, Dec** | Peak conversion months | Schedule intensive campaigns in these months. |
| 6 | **Month: May** | Highest volume, below-average conversion | Reduce May call volume; reallocate to Sep/Oct. |
| 7 | **Campaign contacts > 3** | Conversion drops sharply after 3 calls | Set a 3-contact cap per client per campaign. |
| 8 | **Housing loan = yes** | Below baseline conversion | Deprioritise this segment or adjust product messaging. |
| 9 | **Contact: cellular** | Higher conversion than telephone | Prefer mobile contact where available. |
| 10 | **`duration`** | Strong correlation with `y` but **leakage** | **Must exclude from any model feature set.** |
