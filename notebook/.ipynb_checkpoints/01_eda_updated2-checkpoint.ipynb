# Bank Marketing Campaign
---
**Dataset:** [UCI Bank Marketing Dataset](https://archive.ics.uci.edu/dataset/222/bank+marketing)
**Business Problem:** A Portuguese bank ran direct phone-call marketing campaigns to sell term deposits.  
The goal of this EDA is to understand *which customer segments convert*, *when to call*, and *how prior contact history affects outcomes* — so that future campaigns can be better targeted.

| Feature   | Description                                                                                                                     | Type        |
|-----------|---------------------------------------------------------------------------------------------------------------------------------|-------------|
| `age`     | Age of the client (in years)                                                                                                    | Numeric     |
| `job`     | Type of job (admin., blue-collar, technician, management, services, retired, self-employed, entrepreneur, unemployed, housemaid, student, unknown) | Categorical |
| `marital` | Marital status (married, single, divorced). "Divorced" includes widowed                                                         | Categorical |
| `education` | Education level (primary, secondary, tertiary, unknown)                                                                       | Categorical |
| `default` | Has credit in default? (yes/no)                                                                                                 | Binary      |
| `balance` | Average yearly account balance (in euros)                                                                                       | Numeric     |
| `housing` | Has a housing loan? (yes/no)                                                                                                    | Binary      |
| `loan`    | Has a personal loan? (yes/no)                                                                                                   | Binary      |

## Last Contact of the Current Campaign

| Feature    | Description                                                | Type        |
|------------|------------------------------------------------------------|-------------|
| `contact`  | Contact communication type (cellular, telephone, unknown)  | Categorical |
| `day`      | Day of the month of the last contact (1-31)                | Numeric     |
| `month`    | Month of the last contact (jan-dec)                        | Categorical |
| `duration` | Duration of the last call, in seconds                      | Numeric     |

## Other Campaign Information

| Feature    | Description                                                                                        | Type        |
|------------|----------------------------------------------------------------------------------------------------|-------------|
| `campaign` | Number of contacts made to this client during the current campaign (includes the last contact)     | Numeric     |
| `pdays`    | Days since the client was last contacted in a previous campaign (`-1` = not previously contacted)  | Numeric     |
| `previous` | Number of contacts made to this client before the current campaign                                 | Numeric     |
| `poutcome` | Outcome of the previous marketing campaign (success, failure, other, unknown)                      | Categorical |

## Target Variable

| Feature | Description                                         | Type   |
|---------|-----------------------------------------------------|--------|
| `y`     | Did the client subscribe to a term deposit? (yes/no) | Binary |

---
## How this notebook is organised (the analyst workflow)

| Step | What we do | Why |
|------|-----------|-----|
| 1. Business question | Which customers say *yes*, when, and after how many calls? | Every chart / test below must help answer this |
| 2. Data audit | Missing values, duplicates, placeholders (`unknown`, `-1`), impossible values | Wrong data gives wrong conclusions |
| 3. Baseline | Overall conversion rate with a confidence interval | A segment is only "good" if it beats the baseline |
| 4. Univariate analysis | Distribution of every column (shape, skew, outliers) | Decides which tests are valid (e.g. normal or not) |
| 5. Bivariate analysis + hypothesis tests | Each feature vs target, with a statistical test **and** an effect size | p-value says "is there a difference", effect size says "does it matter" |
| 6. Multivariate analysis | Correlation, segments, logistic regression | Checks if an effect survives after controlling for other features |
| 7. Conclusions | Findings, business actions, limitations | The actual deliverable |

**Which statistical test is used where**

| Question | Test used | Main assumptions |
|----------|-----------|------------------|
| Is a categorical feature related to `y`? (job, month, ...) | Chi-square test of independence + Cramér's V | Independent rows, expected count >= 5 in every cell |
| Do 2 specific groups convert differently? (cellular vs telephone) | Two-proportion z-test | Independent groups, large samples |
| Does a numeric feature differ between yes / no? (balance, age) | Mann-Whitney U + rank-biserial r (Welch t-test shown for comparison) | Independent groups; no normality needed |
| Is a numeric feature normal? | Shapiro-Wilk + Q-Q plot | Only a guide: with 45,000 rows even tiny deviations are "significant" |
| Do features move together? | Spearman correlation | Monotonic relationship; works for skewed data |
| Does an effect remain after controlling for other features? | Logistic regression (odds ratios) | Independent rows, no strong multicollinearity, enough events per feature |

**Rules followed throughout**
* Significance level alpha = 0.05. Because we run many tests, we also report a **Bonferroni-adjusted alpha** (0.05 / number of tests) at the end.
* With 45,211 rows almost everything becomes "statistically significant". So we always read the **effect size** (Cramér's V: ~0.1 small, ~0.3 medium, ~0.5 large) next to the p-value.
* All conversion rates are shown with a **95% confidence interval**, because small segments (e.g. December, only 214 calls) are very uncertain.
* This is an *observational* dataset, so we describe **association**, not cause and effect.

## 1. Environment & Imports <a id='1'></a>
We import the libraries, fix the random seed (so sample-based results repeat), and set the significance level `ALPHA = 0.05` once for the whole notebook.

```python
# Basic libraries for data handling and plotting
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Statistical tests
from scipy.stats import chi2_contingency, mannwhitneyu, ttest_ind, shapiro, levene, probplot, spearmanr
from statsmodels.stats.proportion import proportions_ztest, proportion_confint

# Logistic regression (used once, in section 8.4)
import statsmodels.formula.api as smf

RANDOM_SEED = 42
np.random.seed(RANDOM_SEED)

# Significance level used for every test in this notebook
ALPHA = 0.05

# Plot settings
sns.set_theme(style='whitegrid', palette='muted')
plt.rcParams.update({'figure.dpi': 100, 'figure.figsize': (10, 5),
                     'axes.titlesize': 14, 'axes.labelsize': 12})

# We store every test result in this list and show one summary table at the end
test_log = []
```

**Result:** Libraries loaded. `test_log` is an empty list - every hypothesis test we run will add one row to it, so we can build one summary table at the end.

**Note:** The original notebook used `warnings.filterwarnings('ignore')`. This was removed because hiding warnings can hide real problems.

```python
# Small helper functions so we do not repeat the same code again and again

def conv_table(data, col):
    """Conversion rate (%) for every category of a column, with 95% Wilson confidence interval."""
    t = data.groupby(col, observed=True)['y_num'].agg(total='count', converted='sum').reset_index()
    t['conv_rate'] = t['converted'] / t['total'] * 100
    ci = [proportion_confint(c, n, alpha=ALPHA, method='wilson') for c, n in zip(t['converted'], t['total'])]
    t['ci_low'] = [c[0] * 100 for c in ci]
    t['ci_high'] = [c[1] * 100 for c in ci]
    return t


def chi_square_test(data, col, label=None):
    """Chi-square test of independence between a categorical column and the target.
    Also returns Cramer's V (effect size) and the smallest expected count (assumption check)."""
    table = pd.crosstab(data[col], data['y'])
    chi2, p, dof, expected = chi2_contingency(table)
    cramers_v = np.sqrt(chi2 / (table.values.sum() * (min(table.shape) - 1)))
    test_log.append({'feature': label or col, 'test': 'Chi-square', 'statistic': chi2,
                     'p_value': p, 'effect_size': cramers_v, 'effect_name': "Cramer's V"})
    print(f"Chi-square test ({col} vs y): chi2 = {chi2:.2f}, dof = {dof}, p = {p:.2e}")
    print(f"Cramer's V (effect size) = {cramers_v:.3f}")
    print(f"Smallest expected count  = {expected.min():.1f}  (should be >= 5 for the test to be valid)")


def numeric_vs_target(data, col):
    """Compare a numeric column between y = yes and y = no.
    Mann-Whitney U is the main test (no normality needed). Welch t-test is shown for comparison."""
    yes = data.loc[data['y'] == 'yes', col]
    no = data.loc[data['y'] == 'no', col]
    u, p_mw = mannwhitneyu(yes, no, alternative='two-sided')
    t, p_t = ttest_ind(yes, no, equal_var=False)
    # rank-biserial correlation: effect size for Mann-Whitney (0 = no effect, +-1 = complete separation)
    rbc = 2 * u / (len(yes) * len(no)) - 1
    print(f"{col}: median yes = {yes.median():.0f}, median no = {no.median():.0f} | mean yes = {yes.mean():.1f}, mean no = {no.mean():.1f}")
    print(f"Mann-Whitney U : p = {p_mw:.2e}, rank-biserial r = {rbc:.3f}")
    print(f"Welch t-test   : t = {t:.2f}, p = {p_t:.2e}")
    test_log.append({'feature': col, 'test': 'Mann-Whitney U', 'statistic': u,
                     'p_value': p_mw, 'effect_size': rbc, 'effect_name': 'rank-biserial r'})


def two_group_ztest(data, col, group_a, group_b):
    """Two-proportion z-test: is the conversion rate of group_a different from group_b?"""
    a = data[data[col] == group_a]['y_num']
    b = data[data[col] == group_b]['y_num']
    z, p = proportions_ztest([a.sum(), b.sum()], [len(a), len(b)])
    print(f"{col}: {group_a} = {a.mean()*100:.1f}% (n={len(a):,})  vs  {group_b} = {b.mean()*100:.1f}% (n={len(b):,})  ->  z = {z:.2f}, p = {p:.2e}")


def bar_with_ci(ax, t, xcol, horizontal=False, baseline=None, title=''):
    """Bar chart of conversion rate with 95% CI error bars and a dashed baseline."""
    err = [t['conv_rate'] - t['ci_low'], t['ci_high'] - t['conv_rate']]
    if horizontal:
        ax.barh(t[xcol].astype(str), t['conv_rate'], xerr=err, color='steelblue', edgecolor='white', capsize=3)
        if baseline is not None:
            ax.axvline(baseline, color='red', linestyle='--', label=f'Baseline {baseline:.1f}%')
        ax.set_xlabel('Conversion Rate (%)')
    else:
        ax.bar(t[xcol].astype(str), t['conv_rate'], yerr=err, color='steelblue', edgecolor='white', capsize=3)
        if baseline is not None:
            ax.axhline(baseline, color='red', linestyle='--', label=f'Baseline {baseline:.1f}%')
        ax.set_ylabel('Conversion Rate (%)')
    ax.set_title(title)
    ax.legend()
```

**What this cell does:** defines 5 small reusable functions, so each later cell is short and every feature is tested in exactly the same way.

| Function | Purpose |
|----------|---------|
| `conv_table` | Conversion rate per category + 95% Wilson confidence interval |
| `chi_square_test` | Chi-square test, Cramér's V, and the expected-count assumption check |
| `numeric_vs_target` | Mann-Whitney U (main) and Welch t-test (for comparison) for numeric columns |
| `two_group_ztest` | Compares conversion of two specific groups |
| `bar_with_ci` | Bar chart with CI error bars and baseline line |

**Why Wilson interval?** The simple "p ± 1.96·SE" interval is poor for small groups or rates near 0% / 100%. Wilson works well in both cases.

## 2. Data Loading & Schema <a id='2'></a>

```python
# Load the dataset (the file uses ';' as separator)
# NOTE: change DATA_PATH if your csv is in another folder, e.g. "../data/bank-full.csv"
DATA_PATH = "../data/bank-full.csv"
df = pd.read_csv(DATA_PATH, sep=';')

# Numeric version of the target (1 = subscribed, 0 = not). Makes averages = conversion rate.
df['y_num'] = (df['y'] == 'yes').astype(int)

print("Shape:", df.shape)
df.head()
```

**Result:** 45,211 clients and 17 original columns (18 after we add `y_num`).

**Assumption:** each row is a different client call record (no duplicate clients). We cannot verify this because there is no client ID column.

```python
# Column data types
# We use "not number" instead of "object" so the code works on both old and new pandas versions
num_cols = df.select_dtypes(include='number').columns.drop('y_num').tolist()
cat_cols = df.select_dtypes(exclude='number').columns.tolist()

print(df.dtypes)
print("\nNumerical columns  :", num_cols)
print("Categorical columns:", cat_cols)
```

**Result:** 7 numeric columns and 10 text columns (`y` included).

## 3. Data Quality Audit <a id='3'></a>
We check for: missing values, duplicates, `unknown` placeholders, special codes (`-1`), and impossible values. Decisions are written down so the analysis is reproducible.

```python
# 3a. Missing values and duplicate rows
print("Total missing values :", df.isnull().sum().sum())
print("Duplicate rows       :", df.duplicated().sum())
```

**Key findings:** 0 missing values and 0 duplicate rows.

**Caution:** "no nulls" does not mean "complete" - this dataset stores missing information as the word `unknown` (next cell) and as `-1` (cell after that).

```python
# 3b. 'unknown' is a placeholder for missing information in categorical columns
unknown_report = []
for col in cat_cols:
    n = (df[col] == 'unknown').sum()
    if n > 0:
        unknown_report.append({'column': col, 'unknown_count': n, 'unknown_pct': round(n / len(df) * 100, 2)})

unknown_df = pd.DataFrame(unknown_report).sort_values('unknown_pct', ascending=False)
print("Columns with 'unknown' values:")
print(unknown_df.to_string(index=False))
```

**Key findings:**
* `poutcome` 81.8% unknown, `contact` 28.8%, `education` 4.1%, `job` 0.6%.
* `poutcome = unknown` is not really "missing" - it means the client was **never contacted before** (checked in the next cell).

**Decision:**
* Keep `unknown` as its own category. Dropping rows would remove 82% of the data (poutcome) and the unknown group may behave differently - we will see later that unknown `contact` converts at only 4%.
* Do not impute: `unknown` may be non-random (e.g. missing by campaign design), so guessing values could add bias.

```python
# 3c. pdays = -1 is a special code (client was never contacted before), not a real number of days
never_contacted = (df['pdays'] == -1).sum()
print(f"pdays = -1 (never contacted before): {never_contacted:,} rows ({never_contacted/len(df)*100:.1f}%)")

# Consistency check: pdays and previous should tell the same story
print("Rows with pdays = -1 but previous > 0 :", ((df['pdays'] == -1) & (df['previous'] > 0)).sum())
print("Rows with pdays > 0  but previous = 0 :", ((df['pdays'] > 0) & (df['previous'] == 0)).sum())
print("Rows with poutcome = 'unknown' but client WAS contacted before:",
      ((df['poutcome'] == 'unknown') & (df['pdays'] != -1)).sum())

# New binary column: was the client contacted in an earlier campaign?
df['was_previously_contacted'] = (df['pdays'] != -1).astype(int)

# pdays statistics only for clients who were really contacted before
print("\npdays summary (excluding -1):")
print(df.loc[df['pdays'] != -1, 'pdays'].describe().round(1))
```

**Key findings:**
* 36,954 clients (81.7%) were never contacted before (`pdays = -1`). `pdays` and `previous` are consistent (no contradictions).
* Only **5 rows** have `poutcome = unknown` although they were contacted before - negligible.
* Among the 8,257 previously contacted clients, the median gap since the last contact is 194 days.

**Decision:** `-1` is a code, not a number of days. If we averaged it with real values we would get a meaningless 40 days. So we create `was_previously_contacted` (0/1) and analyse real `pdays` values separately (section 9).

```python
# 3d. Check numeric ranges for impossible values
print(df[num_cols].describe().T[['min', 'max', 'mean', '50%']].round(2))

print("\nCalls with duration = 0 seconds:", (df['duration'] == 0).sum())
print("Clients with negative balance  :", (df['balance'] < 0).sum())
print("Largest value of 'previous'     :", df['previous'].max(), "-> 99th percentile is", round(df['previous'].quantile(0.99), 1))
```

**Key findings:**
* Ages 18-95: realistic.
* 3,766 clients (8.3%) have a negative balance - valid (overdraft).
* Max `balance` = 102,127 EUR vs median 448 EUR - very large values, but possible for rich customers, so **we keep them**.
* 3 calls have `duration = 0` (possibly unanswered calls). Harmless, we keep them.
* `previous` has a maximum of 275 contacts while the 99th percentile is only ~9. That single extreme looks like a data-entry error, but it affects one row and we use rank-based tests, so no action is needed.

**Decision:** no rows are removed. Outliers are handled by *method* (rank-based tests, capping/log in plots and model), not by deleting data.

## 4. Class Imbalance - Target Distribution <a id='4'></a>
> Every conversion rate computed below must be compared with the **baseline conversion rate**.

```python
# Target distribution and the baseline conversion rate
target_counts = df['y'].value_counts()
target_pct = df['y'].value_counts(normalize=True) * 100
BASELINE_CONVERSION = target_pct['yes']

# 95% confidence interval for the baseline (Wilson method)
low, high = proportion_confint(target_counts['yes'], len(df), alpha=ALPHA, method='wilson')

print("=== Target Distribution ===")
for label in ['yes', 'no']:
    print(f"  {label:>3}: {target_counts[label]:>6,}  ({target_pct[label]:.1f}%)")
print(f"\nBaseline conversion rate: {BASELINE_CONVERSION:.2f}%  (95% CI: {low*100:.2f}% to {high*100:.2f}%)")
print("Imbalance ratio (No : Yes) =", round(target_counts['no'] / target_counts['yes'], 1), ": 1")
print("A model that always predicts 'no' would have accuracy =", round(target_pct['no'], 1), "%")
```

**Key findings:**
* Baseline conversion = **11.70%** (95% CI 11.41% - 12.00%): about 1 in 9 calls succeeds.
* Imbalance is 7.5 : 1 (No : Yes).
* A "dumb" model that always says *no* would be **88.3% accurate** while finding zero buyers.

**Implication:** accuracy is a misleading metric. For modelling use ROC-AUC, PR-AUC or F1 on the *yes* class, with stratified train/test split and class weights.

**Assumption:** the dataset represents calls in general. If the bank only called "promising" clients, 11.7% would be higher than the true conversion of the whole customer base.

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 4))
colors = ['#e74c3c', '#2ecc71']

axes[0].bar(target_counts.index, target_counts.values, color=colors, edgecolor='white')
axes[0].set_title('Target Class Distribution')
axes[0].set_ylabel('Number of Clients')
axes[0].set_xlabel('Subscribed Term Deposit (y)')

axes[1].pie(target_counts.values, labels=target_counts.index, autopct='%1.1f%%',
            colors=colors, startangle=90, wedgeprops={'edgecolor': 'white', 'linewidth': 2})
axes[1].set_title('Class Split')

plt.suptitle('Class Imbalance: ~88% No vs ~12% Yes', fontsize=13)
plt.tight_layout()
plt.show()
```

**Reading the chart:** the green (yes) bar is about one-eighth of the red (no) bar. Imbalance is clear but not extreme (not 99:1), so normal methods still work if we use the right metrics.

## 5. Univariate Analysis - Numerical Features <a id='5'></a>
Goal: understand the **shape** of each numeric column, because the shape decides which tests and transformations are valid.

### 5.1 Age

```python
print(df['age'].describe().round(2))
print(f"Skewness : {df['age'].skew():.3f}")
print(f"Kurtosis : {df['age'].kurtosis():.3f}")

# Shapiro-Wilk normality test (on a random sample of 5000, because the test is very sensitive for large data)
stat, p = shapiro(df['age'].sample(5000, random_state=RANDOM_SEED))
print(f"\nShapiro-Wilk on 5000 sampled ages: W = {stat:.3f}, p = {p:.2e}")

# Outliers using the IQR rule
Q1, Q3 = df['age'].quantile([0.25, 0.75])
IQR = Q3 - Q1
lower, upper = Q1 - 1.5 * IQR, Q3 + 1.5 * IQR
n_out = ((df['age'] < lower) | (df['age'] > upper)).sum()
print(f"IQR fence: [{lower:.0f}, {upper:.0f}] | Outliers: {n_out:,} ({n_out/len(df)*100:.1f}%)")
```

**Key findings:**
* Mean 40.9, median 39, range 18-95. Skewness 0.69 (moderate right skew), kurtosis 0.32.
* **Shapiro-Wilk p ≈ 2e-34 -> age is NOT perfectly normal.**
* 487 clients (1.1%) are above the IQR fence of 70 years - these are real elderly clients, not errors.

**Assumption check:** the original comment said "age is roughly normally distributed". That is only approximately true: it looks bell-shaped, but with 45,000 rows even a small deviation is detected by Shapiro-Wilk, and the right tail is longer. So for age we prefer **rank-based tests** or **age bands** instead of tests that need normality.

**Note:** Shapiro-Wilk is run on 5,000 random rows because it becomes over-sensitive (and scipy gives a warning) above that size.

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

sns.histplot(df['age'], bins=30, kde=True, stat='density', color='steelblue', ax=axes[0])
axes[0].set_title('Age Distribution')

sns.boxplot(y=df['age'], color='steelblue', ax=axes[1])
axes[1].set_title('Age Boxplot')

probplot(df['age'], plot=axes[2])
axes[2].set_title('Age Q-Q Plot')

plt.tight_layout()
plt.show()
```

**Reading the charts:** the histogram peaks around age 32, falls sharply after ~60 (probably retirement age) and has a thin tail up to 95. The Q-Q plot leaves the red line at both ends: the lower end is bounded at 18 and the upper end rises faster than a normal curve (heavier right tail), so it is not normal, which agrees with Shapiro-Wilk. The boxplot shows clients above 70 as separate dots - expected for a bank's customers.

### 5.2 Balance

```python
print(df['balance'].describe().round(2))
print(f"Skewness : {df['balance'].skew():.2f}")

neg = (df['balance'] < 0).sum()
print(f"Negative balances: {neg:,} ({neg/len(df)*100:.1f}%)")

# Outliers using the IQR rule
Q1, Q3 = df['balance'].quantile([0.25, 0.75])
IQR = Q3 - Q1
upper = Q3 + 1.5 * IQR
print(f"Balances above the IQR upper fence ({upper:,.0f}): {(df['balance'] > upper).sum():,}")
```

**Key findings:**
* Skewness **8.4** - extremely right-skewed: mean (1,362 EUR) is three times the median (448 EUR).
* 4,712 clients are above the IQR fence (3,462 EUR). The IQR rule flags 10% of rows because the distribution is so skewed - these are not "errors".

**Decision:**
* Do not delete high balances (they are real customers, and may be exactly the valuable ones).
* For testing, use **median-based / rank-based** methods (Mann-Whitney) instead of comparing means.
* For modelling, cap at the 99th percentile or use log transform.

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 4))

# Left: clip at 99th percentile only to make the plot readable (data itself is not changed)
clip_val = df['balance'].quantile(0.99)
axes[0].hist(df['balance'].clip(upper=clip_val), bins=50, color='steelblue', edgecolor='white')
axes[0].set_title(f'Balance (clipped at 99th pct = {clip_val:,.0f})')
axes[0].set_xlabel('Balance (EUR)')

# Right: log scale for positive balances
bal_pos = df.loc[df['balance'] > 0, 'balance']
axes[1].hist(np.log1p(bal_pos), bins=50, color='steelblue', edgecolor='white')
axes[1].set_title('log(1 + Balance) - positive balances only')
axes[1].set_xlabel('log(1 + Balance)')

plt.tight_layout()
plt.show()
```

**Reading the charts:** after clipping (left) most clients are below ~1,500 EUR. After a log transform (right) the shape is much closer to symmetric, which is why log is recommended for models that assume a linear relationship.

**Note:** log cannot be taken for zero / negative balances (about 8% of rows are negative plus some zeros). A signed-log, binning (we use quartiles in 7.5) or capping are the alternatives.

### 5.3 Duration

```python
print(df['duration'].describe().round(2))

fig, ax = plt.subplots(figsize=(10, 4))
ax.hist(df['duration'].clip(upper=df['duration'].quantile(0.99)), bins=50, color='salmon', edgecolor='white')
ax.set_title('Call Duration Distribution (clipped at 99th percentile)')
ax.set_xlabel('Duration (seconds)')
plt.tight_layout()
plt.show()

# Why we must NOT use duration as a predictor: it is only known after the call ends
numeric_vs_target(df, 'duration')
```

**Key findings:**
* Median call = 180 s, but the mean is 258 s because of a long right tail (max 4,918 s).
* Calls that ended in **yes** are much longer: median 426 s vs 164 s (Mann-Whitney p ≈ 0, rank-biserial r = 0.62 - by far the strongest effect in the whole dataset).

> **Data leakage warning - `duration` is excluded from all modelling and from the logistic regression below.**
> Call length is only known **after** the call ends. A longer call is partly a *result* of the client being interested, not a cause. Using it to predict who to call would be impossible in real life (we do not know the length before dialling) and would give a falsely excellent model.

**Use of this variable:** it is fine to describe *why* a call succeeded, but not to decide *whom* to call.

### 5.4 Campaign Contacts

```python
print(df['campaign'].describe().round(2))
print(f"\nClients contacted more than 5 times : {(df['campaign'] > 5).sum():,} ({(df['campaign'] > 5).mean()*100:.1f}%)")
print(f"Clients contacted more than 10 times: {(df['campaign'] > 10).sum():,} ({(df['campaign'] > 10).mean()*100:.1f}%)")

fig, ax = plt.subplots(figsize=(10, 4))
ax.hist(df['campaign'].clip(upper=20), bins=20, color='steelblue', edgecolor='white')
ax.set_title('Contacts in this Campaign (clipped at 20)')
ax.set_xlabel('Number of contacts')
plt.tight_layout()
plt.show()
```

**Key findings:**
* Median 2 contacts, mean 2.76, max **63** (extreme). 9.6% of clients were contacted more than 5 times and 2.6% more than 10 times.
* Highly right-skewed, like balance.

**Decision:** cap `campaign` at 10 for plots / models ("10 = 10 or more") so a few extreme clients do not dominate.

**Business note:** more than 10 calls to one client is probably not good customer experience; later we check if it is also not effective.

## 7. Bivariate Analysis - Feature vs Target <a id='7'></a>
Each feature is compared with the target in the same 3 steps: **(1) conversion table + CI, (2) hypothesis test + effect size, (3) chart with the baseline (red dashed line)**.

For each test we state the hypotheses:
* **H₀ (null):** the feature and `y` are independent (conversion is the same in every group).
* **H₁ (alternative):** conversion differs between groups.
* Decision rule: reject H₀ if p < 0.05 (and, with many tests, if p < the Bonferroni alpha reported in section 10).

### 7.1 Age vs Target

```python
# Group age into bands so the result is easy to explain to the business team
df['age_band'] = pd.cut(df['age'], bins=[17, 25, 35, 45, 55, 65, 100],
                        labels=['18-25', '26-35', '36-45', '46-55', '56-65', '66+'])

age_acc = conv_table(df, 'age_band')
print(age_acc.round(1).to_string(index=False))
print()
chi_square_test(df, 'age_band', label='age (in bands)')

fig, axes = plt.subplots(1, 2, figsize=(13, 5))
bar_with_ci(axes[0], age_acc, 'age_band', baseline=BASELINE_CONVERSION, title='Conversion Rate by Age Band (with 95% CI)')
axes[0].set_xlabel('Age Band')

sns.boxplot(x='y', y='age', hue='y', data=df, palette={'yes': '#2ecc71', 'no': '#e74c3c'}, legend=False, ax=axes[1])
axes[1].set_title('Age Distribution by Subscription')
plt.tight_layout()
plt.show()
```

**Key findings:**
* Conversion by age band is **U-shaped**: 24.0% (18-25), 12.0%, 9.4%, 9.4%, 14.1% (56-65), and **42.6% for 66+** (n = 751; CI 39.1% - 46.2%).
* Chi-square: χ² = 1036, dof = 5, p ≈ 1e-221 -> reject H₀; age band and subscription are associated. Cramér's V = 0.15 (small-to-medium effect).
* Smallest expected count = 87.9 (≥ 5), so the chi-square assumption is satisfied.

**Insight:** the youngest (<=25) and oldest (66+) clients are the best segments. People between 36 and 55 are *below* the 11.7% baseline.

**Assumption:** rows are independent. Age bands were chosen by us (not by the data), so the exact cut points are a judgement - but the U-shape is strong enough to be visible with any reasonable cut points.

```python
# Second test: is the AVERAGE age different between yes and no clients?
# (The original notebook used ANOVA here. With only 2 groups, ANOVA = t-test, and it assumes
#  equal variances and normal data. Let us check both assumptions first.)
yes_age = df.loc[df['y'] == 'yes', 'age']
no_age = df.loc[df['y'] == 'no', 'age']

lev_stat, lev_p = levene(yes_age, no_age)
print(f"Levene test (equal variance?): p = {lev_p:.2e}  -> variances are {'DIFFERENT' if lev_p < ALPHA else 'similar'}")
print(f"Std of age: yes = {yes_age.std():.1f}, no = {no_age.std():.1f}\n")

numeric_vs_target(df, 'age')
```

**Key findings:**
* Levene test p ≈ 1e-133: the variance of age is **different** (std 13.5 for yes vs 10.2 for no). The classic ANOVA / t-test assumes **equal** variances - this assumption is violated.
* Mean age: 41.7 (yes) vs 40.8 (no). Medians: 38 vs 39.
* Welch t-test (corrects for unequal variance): p = 1.6e-05 -> "significant".
* **Mann-Whitney U: p = 0.063 -> NOT significant at 5%.** Effect size r = -0.016 (essentially zero).

**Correction of the original notebook:**
1. The original ran a one-way ANOVA (F = 28.63) on two groups. For two groups ANOVA is identical to a t-test, and it needs equal variances - not true here.
2. It then printed "Insight: Seniors (65+) have significantly higher conversion". **ANOVA on mean age cannot say that** - it only compares two overall averages. The statement is true, but the correct evidence is the age-band chi-square test above.
3. The overall difference in age between yes and no is tiny and not robust (it flips between significant / not significant depending on the test). The reason is the **U-shape**: young and old clients both convert more, so their effects cancel in the overall average age.

**Lesson:** a numeric feature can matter a lot even when its mean does not differ. Always check the shape of the relationship (here: bands).

### 7.2 Job vs Target

```python
job_acc = conv_table(df, 'job').sort_values('conv_rate').reset_index(drop=True)
print(job_acc.round(1).to_string(index=False))
print()
chi_square_test(df, 'job')

fig, ax = plt.subplots(figsize=(10, 6))
bar_with_ci(ax, job_acc, 'job', horizontal=True, baseline=BASELINE_CONVERSION, title='Conversion Rate by Job (with 95% CI)')
plt.tight_layout()
plt.show()
```

**Key findings:**
* Best: **student 28.7%** (CI 25.9-31.7%) and **retired 22.8%** (CI 21.1-24.6%) - roughly 2-2.5× the baseline.
* Worst: blue-collar 7.3%, entrepreneur 8.3%, housemaid 8.8%, services 8.9% - all below the baseline.
* Chi-square: χ² = 836, dof = 11, p ≈ 3e-172 -> job and `y` are associated. Cramér's V = 0.14.
* Smallest expected count = 33.7 -> assumption met.

**Caution:**
* Student and retired are exactly the groups at the two ends of the age U-shape. Part of the "job effect" may really be an **age effect**. Section 8.4 (logistic regression) checks this.
* `unknown` job (n = 288) has a very wide CI (8.6-16.0%), so do not read anything into its rate.

### 7.3 Education vs Target

```python
edu_acc = conv_table(df, 'education').sort_values('conv_rate', ascending=False).reset_index(drop=True)
print(edu_acc.round(1).to_string(index=False))
print()
chi_square_test(df, 'education')

# Post-hoc check: tertiary vs secondary and secondary vs primary (Bonferroni: 2 comparisons)
print("\nPairwise z-tests (use alpha = 0.05 / 2 = 0.025):")
two_group_ztest(df, 'education', 'tertiary', 'secondary')
two_group_ztest(df, 'education', 'secondary', 'primary')
```

**Key findings:**
* Conversion rises with education: primary 8.6% -> secondary 10.6% -> tertiary 15.0%. `unknown` (13.6%) sits close to tertiary.
* Chi-square: χ² = 238.9, dof = 3, p = 1.6e-51 -> reject H₀. **Cramér's V = 0.073 - small effect** (very significant p-value, but education explains little on its own).
* Post-hoc z-tests: tertiary > secondary (p ≈ 7e-36) and secondary > primary (p ≈ 3e-6). Both stay significant after Bonferroni correction (alpha 0.025), so the ordering is real.

**Interpretation of the original test:** the original notebook explained χ², dof and p correctly; we add the effect size and the pairwise tests, which tell us *which* groups differ (the chi-square alone only says "something differs").

**Assumption:** `unknown` is treated as a category. It may be a mix of different education levels, so it should not be merged into tertiary just because the rates look similar.

### 7.4 Marital Status vs Target

```python
mar_acc = conv_table(df, 'marital')
print(mar_acc.round(1).to_string(index=False))
print()
chi_square_test(df, 'marital')

# Post-hoc pairwise tests (Bonferroni: 3 comparisons -> alpha = 0.05 / 3 = 0.0167)
print("\nPairwise z-tests (use alpha = 0.05 / 3 = 0.0167):")
two_group_ztest(df, 'marital', 'single', 'married')
two_group_ztest(df, 'marital', 'divorced', 'married')
two_group_ztest(df, 'marital', 'single', 'divorced')

fig, ax = plt.subplots(figsize=(7, 4))
bar_with_ci(ax, mar_acc, 'marital', baseline=BASELINE_CONVERSION, title='Conversion Rate by Marital Status (with 95% CI)')
plt.tight_layout()
plt.show()
```

**Key findings:**
* Single 14.9% > divorced 11.9% > married 10.1%.
* Chi-square: χ² = 196.5, dof = 2, p = 2e-43; Cramér's V = 0.066 (small).
* All 3 pairwise z-tests are significant even with Bonferroni (alpha 0.0167): single vs married p ≈ 1e-44, divorced vs married p ≈ 8e-5, single vs divorced p ≈ 1.5e-7.

**Caution:** marital status is related to age (singles are younger). Some of this effect may come from age - checked in 8.4.

**Missing in the original:** this section had a chart but no statistical test, and no confidence interval.

### 7.5 Balance vs Target

```python
# Numeric test first
numeric_vs_target(df, 'balance')

# Then balance quartiles: gives a business-friendly view of the same relationship
df['balance_quartile'] = pd.qcut(df['balance'], 4, labels=['Q1 (lowest)', 'Q2', 'Q3', 'Q4 (highest)'])
bal_acc = conv_table(df, 'balance_quartile')
print()
print(bal_acc.round(1).to_string(index=False))
print()
chi_square_test(df, 'balance_quartile', label='balance (in quartiles)')

fig, axes = plt.subplots(1, 2, figsize=(13, 5))
clip_val = df['balance'].quantile(0.99)
sns.boxplot(x='y', y='balance', hue='y', data=df[df['balance'] < clip_val],
            palette={'yes': '#2ecc71', 'no': '#e74c3c'}, legend=False, ax=axes[0])
axes[0].set_title('Balance by Subscription (below 99th pct)')
bar_with_ci(axes[1], bal_acc, 'balance_quartile', baseline=BASELINE_CONVERSION, title='Conversion Rate by Balance Quartile')
plt.tight_layout()
plt.show()
```

**Key findings:**
* Clients who subscribe have higher balances: median 733 vs 417 EUR (mean 1,804 vs 1,304).
* **Mann-Whitney p ≈ 7e-101, rank-biserial r = 0.18** (small-to-medium effect). Welch t-test agrees (p ≈ 4e-23).
* Quartile view: conversion rises steadily from **7.2% (lowest 25% balances) to 16.2% (highest 25%)** - a clear monotonic trend. Chi-square on quartiles: p ≈ 4e-97, V = 0.10.

**Why Mann-Whitney and not t-test:** balance has skewness 8.4 and extreme outliers. The t-test compares means, which a few huge balances can pull strongly; Mann-Whitney compares ranks and is not affected. Both agree here, so the conclusion is robust.

**Assumption:** the two groups are independent samples. The quartile bins are equal-sized by construction, so no group is small.

### 7.6 Housing Loan, Personal Loan & Default vs Target

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4.5), sharey=True)   # same y-axis so bars are comparable

for ax, col in zip(axes, ['housing', 'loan', 'default']):
    acc = conv_table(df, col)
    bar_with_ci(ax, acc, col, baseline=BASELINE_CONVERSION, title=f'Conversion by {col}')
    print(acc.round(1).to_string(index=False))
    chi_square_test(df, col)
    # With only two groups, the z-test gives the same answer as the chi-square test
    two_group_ztest(df, col, 'yes', 'no')
    print('-' * 70)

plt.tight_layout()
plt.show()
```

**Key findings:**

* **Personal loan:** no 12.7% vs yes 6.7% (p ≈ 2e-47, V = 0.07).
* **Default:** no 11.8% vs yes 6.4% (p ≈ 2e-6, V = 0.02 - negligible; only 815 clients, CI 4.9-8.3%).

**Interpretation (a hypothesis, not proven):** clients who already have loan payments have less free cash to put into a term deposit.

**Note:** for 2×2 tables the chi-square test and the two-proportion z-test answer the same question and give almost identical p-values (the small difference comes from the continuity correction that `chi2_contingency` applies to 2×2 tables). Both are shown only to demonstrate this.

**Other limitation:** there is no year column, so seasonality cannot be separated from different campaign waves (the UCI documentation says the calls span about 2008-2010).

## 8. Multivariate Analysis <a id='8'></a>
So far each feature was studied alone. Here we check whether the effects **survive when features are considered together**.

### 8.1 Correlation - Numerical Features

```python
# Spearman correlation is used because balance, duration, campaign, pdays and previous are highly skewed
corr_cols = ['age', 'balance', 'day', 'duration', 'campaign', 'pdays', 'previous', 'y_num']
corr = df[corr_cols].corr(method='spearman')

fig, ax = plt.subplots(figsize=(9, 7))
mask = np.triu(np.ones_like(corr, dtype=bool))
sns.heatmap(corr, mask=mask, annot=True, fmt='.2f', cmap='coolwarm', center=0, linewidths=0.5, ax=ax)
ax.set_title('Spearman Correlation Matrix')
plt.tight_layout()
plt.show()

# Correlation of each feature with the target, with p-values
print("Spearman correlation with y (1 = yes):")
for col in ['age', 'balance', 'day', 'duration', 'campaign', 'pdays', 'previous']:
    r, p = spearmanr(df[col], df['y_num'])
    print(f"  {col:<9} r = {r:+.3f}   p = {p:.2e}")
```

**Key findings (Spearman):**
* With the target: `duration` +0.34 (leakage, ignore), `previous` +0.17, `pdays` +0.15, `balance` +0.10, `campaign` -0.08, `day` -0.03, `age` -0.01 (not significant, p = 0.063).
* All correlations with `y` are weak (< 0.2) apart from the leaking `duration`. This is expected for a binary target and non-linear effects (e.g. the U-shape of age).
* `pdays` and `previous` are almost the same variable (Spearman ≈ 0.99, both are "not contacted" for the same 82% of clients) - **do not put both in a linear model** (multicollinearity).

**Change from the original notebook:** it used Pearson correlation. Pearson assumes linear relationships and is distorted by the heavy skew and outliers in balance, campaign, previous. Spearman is based on ranks and is the safer choice. Also, `y` is now in the matrix, so the target relationship can be seen directly.

**Assumption:** Spearman measures monotonic relationships only. Age shows ~0 correlation even though it has a strong U-shaped effect - a reminder that "no correlation" does not mean "no relationship".

### 8.2 Segments - Job × Education

```python
seg = df.groupby(['job', 'education'])['y_num'].agg(total='count', conv_rate='mean').reset_index()
seg['conv_rate'] = seg['conv_rate'] * 100

# Hide combinations with fewer than 30 clients - their rates are too unstable to trust
MIN_SIZE = 30
print("Job x education combinations with fewer than", MIN_SIZE, "clients:", (seg['total'] < MIN_SIZE).sum(), "of", len(seg))
seg.loc[seg['total'] < MIN_SIZE, 'conv_rate'] = np.nan

pivot = seg.pivot(index='job', columns='education', values='conv_rate').round(1)

fig, ax = plt.subplots(figsize=(10, 7))
sns.heatmap(pivot, annot=True, fmt='.1f', cmap='YlGn', linewidths=0.5, ax=ax, cbar_kws={'label': 'Conversion Rate (%)'})
ax.set_title('Conversion Rate (%) - Job x Education (blank = fewer than 30 clients)')
plt.tight_layout()
plt.show()

print("Top 5 segments (with at least 100 clients):")
print(seg[seg['total'] >= 100].sort_values('conv_rate', ascending=False).head(5).round(1).to_string(index=False))
```

**Key findings:**
* Only 1 of the 48 combinations has fewer than 30 clients and is blanked out. (Original notebook showed every cell, including very unstable ones.)
* Among groups with at least 100 clients, the best are **students (secondary 29.7%, tertiary 26.5%)** and **retired (tertiary 27.6%)**.
* **Education matters inside most jobs:** for admin., blue-collar, self-employed and technician, tertiary clients convert about 6-12 points higher than primary clients (e.g. blue-collar 5.8% -> 16.1%).
* **Student and retired are high at every education level** (21-36%). For students the pattern even reverses (primary 36.4% vs tertiary 26.5%), but these cells are small, so do not over-read it.

**Interpretation:** both job and education carry information. The student/retired effect is strong at all education levels, which fits the age explanation found in 8.4.

**Caution:** this is a descriptive table - small cells (e.g. 100-200 clients) still have a CI of about ±6-8 points, so use it to form hypotheses, not for final targeting rules.

### 8.4 Logistic Regression - effects after controlling for other features
Many of the earlier findings may overlap (age ↔ job ↔ marital, month ↔ contact). A logistic regression estimates the effect of each feature **while holding the others fixed**. The result is an **odds ratio (OR)**: OR > 1 means higher odds of subscribing than the reference group, OR < 1 means lower, OR = 1 means no effect.

*This is used for **explanation** (what matters), not for prediction - no train/test split or performance claims here.*

```python
# Logistic regression to see the effect of each feature while holding the others fixed.
# We do NOT use duration (leakage). We also use poutcome instead of pdays/previous
# because they carry almost the same information (avoids multicollinearity).
model_df = df.copy()
model_df['campaign_capped'] = model_df['campaign'].clip(upper=10)            # reduce effect of extreme values
model_df['balance_k'] = model_df['balance'].clip(upper=model_df['balance'].quantile(0.99)) / 1000   # balance in thousand EUR

# Reference (base) category for each categorical variable is given inside Treatment('...')
formula = ("y_num ~ C(age_band, Treatment('36-45')) + C(job, Treatment('admin.')) + C(marital, Treatment('married'))"
           " + C(education, Treatment('secondary')) + C(default, Treatment('no')) + balance_k"
           " + C(housing, Treatment('no')) + C(loan, Treatment('no')) + C(contact, Treatment('cellular'))"
           " + C(month, Treatment('may')) + campaign_capped + C(poutcome, Treatment('unknown'))")

logit_model = smf.logit(formula, data=model_df).fit(disp=0)
print(f"Observations: {int(logit_model.nobs):,} | Pseudo R-squared: {logit_model.prsquared:.3f} | Converged: {logit_model.mle_retvals['converged']}")

# Convert coefficients to odds ratios with 95% confidence intervals
or_table = pd.DataFrame({'odds_ratio': np.exp(logit_model.params),
                         'ci_low': np.exp(logit_model.conf_int()[0]),
                         'ci_high': np.exp(logit_model.conf_int()[1]),
                         'p_value': logit_model.pvalues}).drop('Intercept')

# Make names shorter: C(job, Treatment('admin.'))[T.retired] -> job: retired
or_table.index = [i.replace("C(", "").replace(", Treatment(", " vs ").replace("))[T.", ": ").replace("]", "").replace("'", "")
                  for i in or_table.index]
print(or_table.round(3).sort_values('odds_ratio', ascending=False).to_string())
```

**Model check:** 45,211 rows, model converged, pseudo R² = 0.171 (the model explains a moderate share of the variation - 0.2-0.4 is already considered a very good fit for logit models).

**Key findings (adjusted odds ratios, reference group in brackets):**

| Feature | OR (95% CI) | Meaning |
|---------|-------------|---------|
| poutcome = success (unknown) | **8.8** (7.8-9.9) | Previous success is the strongest signal even after controlling for everything else |
| month = March (May) | **4.6** (3.7-5.7) | Sep, Oct, Dec also ≈ 2.9 |
| age 66+ (36-45) | **2.2** (1.8-2.8) | Seniors really are different |
| age 18-25 (36-45) | **1.8** (1.5-2.2) | So are the very young |
| marital = single (married) | 1.27 (1.17-1.38) | Small effect, remains after controlling age |
| housing loan = yes (no) | **0.60** (0.56-0.65) | 40% lower odds |
| personal loan = yes (no) | 0.68 (0.62-0.76) | |
| contact = unknown (cellular) | **0.27** (0.24-0.31) | Strongly lower, even after controlling for month |
| contact = telephone (cellular) | 0.74 (0.65-0.84) | Lower than cellular once month is controlled |
| campaign (each extra contact) | 0.91 (0.89-0.92) | ~9% lower odds per additional call |

**Findings that changed after adjustment:**
* **Job effects mostly disappear.** Student (OR 1.13, p = 0.23) and retired (OR 1.10, p = 0.32) are no longer significant: the high rates in section 7.2 were mainly an **age effect** (retired people are old, students are young). The original conclusion "target students and retirees" should be rephrased as "target the **young (≤25)** and the **66+**".
* **Default** (OR 0.92, p = 0.58) and most job categories are not significant.
* Balance has a very small effect per 1,000 EUR (OR 1.04).

**Assumptions & limits:**
* Rows independent; relationships are linear in the log-odds (age is handled with bands because of its U-shape).
* No strong multicollinearity: we use `poutcome` instead of `pdays` and `previous`, and exclude `duration` (leakage).
* No interactions were tested (e.g. age × month), and the 12 months / 12 jobs create many comparisons - a few borderline p-values (0.01-0.05) could be chance.

```python
# Forest plot: odds ratio > 1 means higher chance of subscribing (compared with the reference group)
plot_or = or_table.sort_values('odds_ratio')
colors = ['#2ecc71' if (lo > 1) else '#e74c3c' if (hi < 1) else 'grey'
          for lo, hi in zip(plot_or['ci_low'], plot_or['ci_high'])]

fig, ax = plt.subplots(figsize=(10, 12))
ax.errorbar(plot_or['odds_ratio'], range(len(plot_or)),
            xerr=[plot_or['odds_ratio'] - plot_or['ci_low'], plot_or['ci_high'] - plot_or['odds_ratio']],
            fmt='none', ecolor='lightgrey', capsize=2)
ax.scatter(plot_or['odds_ratio'], range(len(plot_or)), c=colors, zorder=3)
ax.axvline(1, color='black', linestyle='--')
ax.set_xscale('log')
ax.set_yticks(range(len(plot_or)))
ax.set_yticklabels(plot_or.index, fontsize=8)
ax.set_xlabel('Odds ratio (log scale)  -  green = higher odds, red = lower odds, grey = not significant')
ax.set_title('Adjusted Odds Ratios (logistic regression, 95% CI)')
plt.tight_layout()
plt.show()
```

**Reading the forest plot:** each dot is an odds ratio, the line is its 95% CI. If the line crosses the vertical dashed line (OR = 1), the effect is not statistically different from the reference group (grey). Green = clearly higher odds, red = clearly lower.

The biggest dots on the right (previous success, March, age 66+) and left (unknown contact, housing loan) are the features worth building campaign rules around.

## 9. Campaign & Contact History <a id='9'></a>

```python
df['campaign_capped'] = df['campaign'].clip(upper=10)
camp_acc = conv_table(df, 'campaign_capped')
print(camp_acc.round(1).to_string(index=False))
print()
chi_square_test(df, 'campaign_capped', label='campaign (capped at 10)')

# The chi-square test only says "different". To test the DIRECTION (more calls -> lower conversion) we use Spearman
r, p = spearmanr(df['campaign'], df['y_num'])
print(f"Spearman (campaign vs y): r = {r:.3f}, p = {p:.2e}")

fig, ax = plt.subplots(figsize=(9, 5))
ax.errorbar(camp_acc['campaign_capped'], camp_acc['conv_rate'],
            yerr=[camp_acc['conv_rate'] - camp_acc['ci_low'], camp_acc['ci_high'] - camp_acc['conv_rate']],
            marker='o', color='steelblue', linewidth=2, capsize=4)
ax.axhline(BASELINE_CONVERSION, color='red', linestyle='--', label=f'Baseline {BASELINE_CONVERSION:.1f}%')
ax.set_title('Conversion Rate vs Number of Contacts (10 = 10 or more)')
ax.set_xlabel('Contacts in this campaign')
ax.set_ylabel('Conversion Rate (%)')
ax.legend()
plt.tight_layout()
plt.show()
```

**Key findings:**
* Conversion falls from **14.6% (1 call)** to 11.2% (2 and 3 calls) and then keeps falling: 9.0% (4), 7.9% (5), 7.1% (6) ... **4.2% (10 or more)**.
* Chi-square p ≈ 9e-70 (V = 0.09). Spearman r = -0.084, p ≈ 1e-71: more calls goes with lower conversion.
* In the regression each extra call reduces the odds by about 9% (OR 0.91) even after controlling for other features.

**Correction of the original notebook:** the original claimed "conversion drops sharply after 3 contacts". The data shows a different pattern: **one drop between call 1 and 2, flat between 2 and 3, and then a steady slow decline**. There is no sharp cliff at 3.

**Important caution (survivorship):** clients who say *yes* usually stop being called, while clients who say *no* keep receiving calls. So high call counts are naturally filled with "no" clients. This does **not** prove that extra calls reduce interest - only that persistent callers rarely convert.

**Business suggestion:** a cap of about 3-5 contacts looks reasonable (rates are below 8% beyond 5 calls and 4% beyond 10). The cap should be tested in a controlled A/B experiment before applying it widely.

```python
prev_acc = conv_table(df, 'was_previously_contacted')
prev_acc['label'] = prev_acc['was_previously_contacted'].map({0: 'Never contacted', 1: 'Previously contacted'})
print(prev_acc.round(1).to_string(index=False))
print()
chi_square_test(df, 'was_previously_contacted', label='previously contacted (yes/no)')

fig, ax = plt.subplots(figsize=(6, 4))
bar_with_ci(ax, prev_acc, 'label', baseline=BASELINE_CONVERSION, title='Previously Contacted vs Not (95% CI)')
plt.tight_layout()
plt.show()
```

**Key findings:**
* Never contacted: 9.2% (n = 36,954). Previously contacted: **23.1%** (n = 8,257) - about 2.5× higher.
* Chi-square: χ² = 1,263, p ≈ 1e-276, V = 0.17.

**Interpretation:** prior contact (even a failed one) identifies a warmer audience. Combined with section 7.9, a re-contact list from earlier campaigns is the cheapest high-yield segment.

```python
# Days since last contact, only for clients who were contacted before
old = df[df['pdays'] != -1].copy()
old['pdays_band'] = pd.cut(old['pdays'], bins=[0, 90, 180, 365, 900], labels=['1-90', '91-180', '181-365', '366+'])
pdays_acc = conv_table(old, 'pdays_band')
print(pdays_acc.round(1).to_string(index=False))
print()
chi_square_test(old, 'pdays_band', label='pdays (bands, previously contacted only)')

fig, axes = plt.subplots(1, 2, figsize=(13, 4))
for label, color in [('no', '#e74c3c'), ('yes', '#2ecc71')]:
    axes[0].hist(old.loc[old['y'] == label, 'pdays'].clip(upper=400), bins=40, alpha=0.5, color=color, label=label, density=True)
axes[0].set_title('Days Since Last Contact (clipped at 400)')
axes[0].set_xlabel('pdays')
axes[0].legend(title='Subscribed')

bar_with_ci(axes[1], pdays_acc, 'pdays_band', baseline=BASELINE_CONVERSION, title='Conversion by Days Since Last Contact')
plt.tight_layout()
plt.show()
```

**Key findings (previously contacted clients only):**
* 1-90 days since last contact: 35.0%; 91-180: 27.4%; 181-365: 17.8%; 366+: 29.2%.
* Chi-square p ≈ 7e-36, V = 0.14.
* Recent contact is better, **but the pattern is not a straight line**: clients last contacted more than a year ago convert better than those contacted 6-12 months ago.

**Interpretation / caution:** we cannot tell why. It may be a real "re-engagement" effect or a different mix of clients in the old campaign. The 366+ group is small (643). Do not turn this into a simple "call within 90 days" rule without more evidence.

## 10. Key Findings, Business Actions & Limitations <a id='10'></a>
### 10.1 All hypothesis tests in one table

```python
# Final summary of all hypothesis tests in this notebook
final = pd.DataFrame(test_log)
alpha_adj = ALPHA / len(final)
final['significant (adj. alpha)'] = np.where(final['p_value'] < alpha_adj, 'Yes', 'No')
final = final.sort_values('effect_size', key=abs, ascending=False)
print(f"Tests run: {len(final)} | Bonferroni-adjusted alpha = {alpha_adj:.4f}")
final[['feature', 'test', 'p_value', 'effect_name', 'effect_size', 'significant (adj. alpha)']].round({'effect_size': 3}).reset_index(drop=True)
```

**How to read it:**
* Every feature except the overall mean age is statistically significant, even after the strict Bonferroni alpha. With this sample size, **p-values alone do not separate important features from minor ones**.
* The **effect size** column does. Ranking by effect size: duration (leakage) > poutcome > month > balance > previously contacted > age bands ≈ contact > housing > job.
* `default` is significant but with a negligible effect (V = 0.02); `age` (as a mean difference) is not significant.

**Note on multiple testing:** with 17 tests, the chance of at least one false positive at alpha = 0.05 is about 58%. That is why the adjusted alpha is used. None of the conclusions above change because the p-values are extremely small.

### 10.4 Limitations
* **Observational data:** all findings are associations. For example, we cannot say calling in March *causes* success.
* **Selection bias:** the bank chose whom to call and when. Rates in small months or segments may reflect who was selected.
* **No year or client ID:** campaign waves and repeated clients cannot be separated; rows may not be fully independent.
* **Large sample:** tiny effects become "significant". Business impact should be judged by effect size and segment size.
* **`unknown` categories** (contact, education, job) are kept as given; imputing might change conclusions.
* **Single dataset, no train/test:** the logistic regression explains this data; it was not validated for prediction.
